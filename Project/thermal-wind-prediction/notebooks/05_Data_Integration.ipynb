```python
# =========================================================
# IMPORTS
# =========================================================

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
# =========================================================
# DATA PATHS
# =========================================================

WIND_PATH = "../data/processed/badalona_wind_target.parquet"

BUOY_PATH = "../data/processed/barcelona_buoy_processed.parquet"

XEMA_PATH = "../data/gold/xema_hourly_model_ready.parquet"
```

```python
# =========================================================
# LOAD DATASETS
# =========================================================

df_wind = pd.read_parquet(WIND_PATH)

df_buoy = pd.read_parquet(BUOY_PATH)

df_xema = pd.read_parquet(XEMA_PATH)

print("Datasets cargados correctamente")
```

```python
# =========================================================
# QUICK DATASET INSPECTION
# =========================================================

datasets = {
    "WIND": df_wind,
    "BUOY": df_buoy,
    "XEMA": df_xema
}

for name, df_ in datasets.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print(f"Shape: {df_.shape}")

    print("\nColumns:")
    print(df_.columns.tolist())

    print("\nDtypes:")
    print(df_.dtypes)

    print("\nFirst rows:")
    display(df_.head(3))
```

```python
# =========================================================
# NORMALIZE DATETIME INDEX
# =========================================================

# ---------------------------------------------------------
# WIND
# ---------------------------------------------------------

df_wind["datetime"] = pd.to_datetime(df_wind["datetime"])

df_wind = df_wind.set_index("datetime")

df_wind = df_wind.sort_index()

# ---------------------------------------------------------
# BUOY
# ---------------------------------------------------------

df_buoy.index = pd.to_datetime(df_buoy.index)

df_buoy = df_buoy.sort_index()

# ---------------------------------------------------------
# XEMA
# ---------------------------------------------------------

df_xema.index = pd.to_datetime(df_xema.index)

df_xema = df_xema.sort_index()

# Limpiar nombres extraños heredados del pivot
df_xema.index.name = "datetime"

df_xema.columns.name = None

# ---------------------------------------------------------
# VALIDATION
# ---------------------------------------------------------

print("NORMALIZATION COMPLETED")
print("-" * 50)

print("\nWIND INDEX:")
print(type(df_wind.index))
print(df_wind.index[:3])

print("\nBUOY INDEX:")
print(type(df_buoy.index))
print(df_buoy.index[:3])

print("\nXEMA INDEX:")
print(type(df_xema.index))
print(df_xema.index[:3])
```

```python
# =========================================================
# WIND → HOURLY RESAMPLE
# =========================================================

df_wind_hourly = (
    df_wind
    .resample("1h")
    .agg({
        "temperature": "mean",
        "wind_direction_deg": "median",
        "wind_avg": "mean",
        "wind_min": "min",
        "wind_max": "max",
        "unknown": "mean"
    })
)

print("WIND HOURLY CREATED")
print("-" * 50)

print(f"Shape: {df_wind_hourly.shape}")

print("\nFirst rows:")
display(df_wind_hourly.head(100))

print("\nMissing values:")
print(df_wind_hourly.isna().sum())
```

```python
# =========================================================
# ANALYZE TEMPORAL GAPS
# =========================================================

wind_diffs = (
    df_wind.index
    .to_series()
    .diff()
)

print("TEMPORAL GAPS ANALYSIS")
print("-" * 50)

print("\nMost common gaps:")
print(
    wind_diffs.value_counts()
    .head(20)
)

print("\nLargest gaps:")
print(
    wind_diffs
    .sort_values(ascending=False)
    .head(20)
)
```

```python
# =========================================================
# CREATE ATMOSPHERIC BASE DATASET
# =========================================================

df_base = df_xema.join(
    df_buoy,
    how="inner"
)

print("BASE DATASET CREATED")
print("-" * 50)

print(f"Shape: {df_base.shape}")

print("\nDatetime range:")
print(df_base.index.min())
print(df_base.index.max())

print("\nMissing values:")
print(
    df_base.isna().sum().sort_values(ascending=False).head(10)
)

print("\nFirst rows:")
display(df_base.head())
```

```python
# =========================================================
# MERGE WIND DATA
# =========================================================

df_master = df_base.join(
    df_wind_hourly,
    how="left"
)

print("MASTER DATASET CREATED")
print("-" * 50)

print(f"Shape: {df_master.shape}")

print("\nDatetime range:")
print(df_master.index.min())
print(df_master.index.max())

print("\nWIND missing values:")
print(
    df_master[
        [
            "temperature",
            "wind_avg",
            "wind_max"
        ]
    ]
    .isna()
    .sum()
)

print("\nFirst rows:")
display(df_master.head())
```

```python
# =========================================================
# INFER OPERATIONAL HOURS FROM DATA
# =========================================================

# ---------------------------------------------------------
# Hours where wind observations exist
# ---------------------------------------------------------

wind_hours = (
    df_master[df_master["wind_avg"].notna()]
    .copy()
)

wind_hours["month"] = wind_hours.index.month
wind_hours["hour"] = wind_hours.index.hour

# ---------------------------------------------------------
# First and last active hour by month
# ---------------------------------------------------------

operational_hours = (
    wind_hours
    .groupby("month")["hour"]
    .agg(["min", "max"])
)

print("INFERRED OPERATIONAL HOURS")
print("-" * 50)

display(operational_hours)
```

```python
# =========================================================
# CREATE DYNAMIC OPERATIONAL WINDOW
# =========================================================

# ---------------------------------------------------------
# Month → operational hours dictionaries
# ---------------------------------------------------------

month_start_hour = operational_hours["min"].to_dict()
month_end_hour = operational_hours["max"].to_dict()

# ---------------------------------------------------------
# Extract temporal features
# ---------------------------------------------------------

df_master["month"] = df_master.index.month
df_master["hour"] = df_master.index.hour

# ---------------------------------------------------------
# Dynamic operational flag
# ---------------------------------------------------------

df_master["is_operational_hour"] = df_master.apply(
    lambda row:
        (
            month_start_hour[row["month"]]
            <= row["hour"]
            <= month_end_hour[row["month"]]
        ),
    axis=1
).astype("int8")

# ---------------------------------------------------------
# Wind data available
# ---------------------------------------------------------

df_master["wind_data_available"] = (
    df_master["wind_avg"]
    .notna()
    .astype("int8")
)

# ---------------------------------------------------------
# Valid wind observation
# ---------------------------------------------------------

df_master["valid_wind_observation"] = (
    (
        df_master["wind_data_available"] == 1
    ) &
    (
        df_master["is_operational_hour"] == 1
    )
).astype("int8")

# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

print("DYNAMIC OPERATIONAL FEATURES CREATED")
print("-" * 50)

print("\nOperational hour distribution:")
print(
    df_master["is_operational_hour"]
    .value_counts()
)

print("\nValid observations:")
print(
    df_master["valid_wind_observation"]
    .value_counts()
)

display(
    df_master[
        [
            "wind_avg",
            "wind_data_available",
            "is_operational_hour",
            "valid_wind_observation"
        ]
    ].head(48)
)
```

```python
# =========================================================
# CREATE OPERATIONAL GAP FEATURE
# =========================================================

df_master["operational_gap"] = (
    (
        df_master["is_operational_hour"] == 1
    ) &
    (
        df_master["wind_data_available"] == 0
    )
).astype("int8")

print("OPERATIONAL GAP FEATURE CREATED")
print("-" * 50)

print(
    df_master["operational_gap"]
    .value_counts()
)

print("\nGap percentage:")

gap_pct = (
    df_master["operational_gap"]
    .mean() * 100
)

print(f"{gap_pct:.2f}%")
```

```python
# =========================================================
# EXPORT MASTER DATASET
# =========================================================

OUTPUT_PATH = "../data/gold/master_hourly_dataset.parquet"

df_master.to_parquet(
    OUTPUT_PATH
)

print("MASTER DATASET EXPORTED")
print("-" * 50)

print(f"Path: {OUTPUT_PATH}")

print(f"\nShape: {df_master.shape}")

print("\nDatetime range:")
print(df_master.index.min())
print(df_master.index.max())

print("\nTotal columns:")
print(len(df_master.columns))

print("\nColumns:")
print(df_master.columns.tolist())
```
