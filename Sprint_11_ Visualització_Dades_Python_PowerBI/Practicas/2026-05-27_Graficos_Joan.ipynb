```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")
```

```python
df.describe()
```

```python
df.info()
```

```python
df.head()
```

```python
# 1 Cat

# sex --> nominal

df['sex'].value_counts()#.plot.pie()
df['sex'].value_counts().plot.pie(startangle = 90)
```

```python
# Day --> ordinal

tabla = df['day'].value_counts()
tabla.plot.bar();

# Si ponemos ; final desaparece el texto que hay arriba
```

```python
# 1 Num

df['tip'].plot.hist(bins = 20);
```

```python
df['tip'].plot.kde();
```

```python
df['tip'].plot.box();
```

```python
# 2 Cat

pd.crosstab(df['sex'], df['day'])
```

```python
pd.crosstab(df['sex'], df['day']).plot.bar();
```

```python
pd.crosstab(df['sex'], df['day']).transpose().plot.bar();
```

```python
# 2 Num

df.plot.scatter(x="total_bill", y = "tip");
```

```python
sns.lmplot(data=df, x="total_bill", y="tip");
```

```python
 # 1 Cat / 1 Num

df.groupby("sex")[['tip']].mean()
```

```python
df.groupby("sex")[['tip']].mean().plot.bar();
```

```python
df.groupby("day")[['tip']].mean().plot.bar();
```

```python
sns.boxplot(df, x="day", y="tip");
```

```python
# > 2 Var
sns.boxplot(df, x="day", y="tip", hue ="sex");
```

```python
sns.violinplot(df, x="day", y="tip", hue ="time", split="sex");
```

```python
sns.lmplot(data=df, x="total_bill", y="tip", hue="sex");
```

```python
sns.lmplot(data=df, x="total_bill", y="tip", hue="sex", col="smoker");
```

```python
sns.lmplot(data=df, x="total_bill", y="tip", hue="sex", col="smoker", row="time");
```

```python
sns.relplot(data=df, x="total_bill", y="tip", hue="sex", col="smoker", row="time", size="size");
```

```python
import plotly.express as px
fig = px.scatter(df, x="total_bill", y="tip", color="sex", symbol="smoker")
fig.show()
```
