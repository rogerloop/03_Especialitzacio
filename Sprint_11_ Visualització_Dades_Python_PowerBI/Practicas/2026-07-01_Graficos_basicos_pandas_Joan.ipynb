```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("penguins")
df.head()
```

```python
df["body_mass_g"].plot.hist(bins = 30);
```

```python
df["body_mass_g"].plot.box();
```

```python
df["species"].value_counts()
```

```python
df["species"].value_counts().plot.bar();
```

```python
df.groupby(["species"])["body_mass_g"].mean()
```

```python
df.groupby(["species"])[["body_mass_g"]].mean().plot.bar();
```

```python
df.groupby(["species"])[["body_mass_g"]].mean().plot.barh();
```

```python
df.groupby(["species"])[["body_mass_g"]].mean().plot.pie(subplots=True);
```

```python
df.groupby(["species"])[["body_mass_g","flipper_length_mm"]].count().plot.pie(subplots=True);
```

```python
df.groupby(["species"])[["body_mass_g","flipper_length_mm"]].mean().plot.bar();
```

```python
df.groupby(["species"])[["body_mass_g","flipper_length_mm"]].mean().plot.bar(stacked=True);
```

```python
# No pita biern con groupby...... mejor usar un crosstab

df.groupby(["species", "sex"])[["body_mass_g","flipper_length_mm"]].mean().plot.bar(stacked=True);
```

```python
pd.crosstab(df["species"], df["island"])
```

```python
pd.crosstab(df["species"], df["island"]).plot.bar()
```

```python
pd.crosstab(df["species"], df["island"]).transpose().plot.bar()
```

```python
pd.crosstab(df["species"], df["island"]).transpose().plot.bar(stacked=True)
```

```python
pd.crosstab(df["species"], df["island"], normalize = "index") #.plot.bar(stacked=True)
```

```python
pd.crosstab(df["species"], df["island"], normalize = "index").plot.bar(stacked=True)
```

```python
df.plot.scatter(x = "flipper_length_mm"	, y = "body_mass_g")
```

```python
df.plot.hexbin(x = "flipper_length_mm"	, y = "body_mass_g", gridsize=(20,20));
```
