### ACCEDER A DATOS

1. ACCESO POR POSICION CON ILOC (poco usado) --> df.iloc [filas, columnas]

```python
df.iloc [0,0] # filas, columnas
# accedo a valor de 1a fila y 1a columna
```

```python
df.iloc [0:3,0:2]
# accedo a 3 primeras filas y 2 1as columnas
```

2. ACCESO POR CONDICION CON .loc

2.1 ACCESO POR NOMBRE (TAMBIEN LLAMADO ACCESO POR ETIQUETA) (MAS COMUN QUE CON EL ILOC, PUES PUEDE CAMBIAR DE NOMBRE LA COLUMNA)

```python
df.loc[0,"sex"]
# Accedo a la fila que se llama "0" y la columna llamada "sex"
```

```python
df.loc[[0,4,8],["sex","day"]]
# coge por nombres de columnas (sex, day) y nombres de filas (0,4,8)
```

2.2 ACCESO POR CONDICIONES --> CONDICION DE FILAS , HABLO DE UNA COLUMNA Y SELECCIONA AQUELLAS FILAS DONDE LA CONDICION COLUMNA DE TIP SEA > 3

```python
df.loc[df["tip"] > 3, ["smoker","time"]] #selecciona las filas con > 3 de propina y ensena solo las columnas smoker y time
df.loc[df["tip"] > 3, :] # me dara todas las filas con tip > 3 y todas las columnas
```

2.2.1 ACCESO CON 2 CONDICIONES TIPO AND Y OR DE PYTHON ES CON & O |

```python
df.loc[(df["tip"] > 3) & (df["tip"] < 5) , ["smoker","time"]]
df.head()
# 2 condiciones tip mayor que 3 y menor que 5 , no fumadores pero solo veo columnas smoker y time
```

```python
df.loc[(df["tip"] > 3) | (df["tip"] < 5) , ["smoker","time"]]
df.head()

# la barra es como el or
```

2.2.3 ACCESO POR COLUMNAS (POCO HABITUAL)

```python
df.loc[(df["tip"] > 3) & (df["tip"] < 5),[c for c in df.columns if "i" in c]]
df.head()
# selecciona las propinas entre 3 y 5 pero solo las columnas que contienen la letra "i"
# Usos en encuestas seleccionar los que tienen un codigo concreto o las columnas que por ejemplo tiene el icono $
```

NOTA IMP:
SI QUEREMOS QUE EL DIA ESTE EN UNA SERIE DE VALORES SE PONE COMO QUE ESOS VALORES ESTEN EN UNA LISTA DE DIAS (ademas aqui de poner condicion de que el nombre de la columna contenga la i)

```python
df.loc[df["day"].isin(["Sun","Sat","Fri"]), [c for c in df.columns if "i" in c]]
df.head()
```
