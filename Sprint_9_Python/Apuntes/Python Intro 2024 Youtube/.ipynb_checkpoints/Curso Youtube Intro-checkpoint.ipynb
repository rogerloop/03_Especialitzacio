# Bucles

Automatizar un proceso

```python
notas = [4,2,3,7,6,8,7,9,3,2,4,5,6,10]
for x in notas:
    print("La nota es:",x)
```

```python
# Bucle por valor
for nota in notas:
    print("\n\nLa nota es:",nota )
    if nota < 5:
        if nota >= 3.5:
            print("Revisaremos el examen")
        else:
            print("Suspenso") # está identado DOS niveles
    else:
        print("Aprobado")
        if nota >= 9:
            print("Matrícula de honor")
```

```python
# Bucles por posición

for posicion in range(len(notas)):
    print("En la posición", posicion, "hay un", notas[posicion])
```

# Creación de funciones propias

```python
def media(numeros, numerodecimales = 2):
    """
    numeros: Lista o similares
    numerodecimales: entero con los decimales (por defecto: 2)
    """
    suma = sum(numeros) # suma
    nnumeros = len(numeros) # cuántos hay
    return round(suma/nnumeros, numerodecimales) #puerta de salida (pueden haber varias, pero cuando llegas a una, fin)
```

```python
media(notas)
```

```python
media([3,3,4,5,6,7,8,9,7,7,55],3)
```

# Tablas resumen

Equivalente a tablas dinámicas de Excel (Pivot table)

```python
### Groupby para hacer tablas por filas
```

```python
df.groupby(["carrier"])[["arr_delay"]].mean().sort_values("arr_delay") # sum / count / min / max / mean / describe
```

```python
df.groupby(["carrier","origin"])[["arr_delay","dep_delay","air_time","distance"]].describe().to_excel("hola.xlsx")
```

```python
### Crosstab para hacer tablas de doble entrada
```
