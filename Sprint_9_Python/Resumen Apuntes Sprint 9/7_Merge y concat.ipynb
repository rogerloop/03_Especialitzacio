```python
import pandas as pd
import numpy as np
```

```python

# 1. Preparación de los datos

df1 = pd.DataFrame({'ID': [1,2,3], 'Nombre': ['Ana', 'Juan', 'Luis']})
df2 = pd.DataFrame({'ID': [3,4 ,5], 'Edad': [25,30,22]})
df3 = pd.DataFrame({'User_ID': [3,4 ,5], 'Edad': [25,30,22]})
df4 = pd.DataFrame({'ID': [1,2,3,3], 'Ciudad': ['Madrid', 'Barcelona', 'Sevilla','Valencia']})
df5 = pd.DataFrame({'Edad': [25,30,22]}, index = [1,2,3])
df6 = pd.DataFrame({'ID': [1,2,3], 'Nombre': ['Carlos', 'Pedro', 'Maria']})
df7 = pd.DataFrame({'ID': [1,2,3,3], 'Edad': [25,30,22,40]})
```

Merge : Mediante nexo / Concat : sin valor de nexo

```python
df1
```

```python
df2
```

```python
# INNER mantiene SOLO las coincidencias
pd.merge(df1,df2, on = 'ID', how = 'inner')
```

```python
# OUTER: mantiene toda la información de ambas tablas, rellenando con vacíos (NaN) donde no hay encaje.
pd.merge(df1,df2, on = 'ID', how = 'outer')
```

```python
#LEFT: mantiene la tabla de la izquierda intacta y añade lo que pueda de la derecha.

left_result = pd.merge(df1, df2, on='ID', how='left')
```

```python
# NO RECOMIENDA
# Right Join (Se recomienda mejor cambiar el orden de las tablas y usar left)
right_result = pd.merge(df1, df2, on='ID', how='right')
```

---
### MERGE NOMBRES DE COLUMNA O INDICES DISTINTOS
----

```python
#Si las columnas de nexo se llaman distinto, usamos `left_on` y `right_on`. También se puede unir usando el índice de la fila con `right_index` o `left_index`.

# Diferentes nombres de columna
pd.merge(df1, df3, left_on='ID', right_on='User_ID')
```

---
### UNIONES CON DIFERENTES NOMBRES DE COLUMNA O INDICES
---

```python
### Uniones con diferentes nombres de columna o índices

# Unir usando el índice de la derecha
pd.merge(df1, df5, left_on = 'ID', right_index=True)
```

---
### MERGE CON SUFIJOS
---

```python
### Parámetros avanzados: Sufijos e Indicador
# Podemos personalizar los sufijos cuando hay columnas duplicadas y usar `indicator` para saber de qué tabla proviene cada registro.

# Uso de sufijos personalizados
pd.merge(df1, df6, on='ID', suffixes=('_izq', '_der'))
```

---
### MERGE CON INDICADOR DE ORIGEN (indicator)


---

```python
# Indicador de origen (útil en outer joins para trazabilidad)
pd.merge(df1, df2, on = 'ID' ,how='outer', indicator=True)
```

### CONCAT
---

```python
## 3. Operación Concat
# concat` se utiliza para **pegar tablas** directamente (ya sea por filas o columnas) sin buscar necesariamente una correspondencia entre valores.

#Concatenar por filas (axis=0) y columnas (axis=1)
#Por defecto, concatena por filas (`axis=0`). Si se desea pegar una al lado de la otra, se usa `axis=1`.


# Concatenar por filas
# Se pasa una lista de DataFrames
pd.concat([df1, df2], axis=0)
```

```python
pd.concat([df1, df2], axis=0)
```

---
### CONCATENAR POR COLUMNAS
---

```python
# Concatenar por columnas
pd.concat([df1, df2], axis=1)
```

---
### INDICES DESALINEADOS
---

```python
### Manejo de índices en Concat
# Para evitar índices repetidos, se usa `ignore_index=True`. También se pueden crear índices múltiples (MultiIndex) para identificar el origen de los datos con `keys`.

# Resetear índices al unir
pd.concat([df1, df2], ignore_index=True)
```

---
### MULTIINDEX
---

```python

# Crear MultiIndex para trazabilidad
pd.concat([df1, df2], keys=['Grupo 1', 'Grupo 2'])

```
