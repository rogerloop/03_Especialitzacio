### Notas sobre los comandos incluidos:
*   **`df.shape`**: Devuelve una tupla donde el primer valor es el número de filas y el segundo el de columnas.
*   **`df.info()`**: Es fundamental para detectar **valores faltantes** y entender si los datos son decimales (`float`), enteros (`int`) u objetos/texto (`object`).
*   **`df.select_dtypes()`**: Permite aislar datos específicos, por ejemplo, para realizar una matriz de correlación (`.corr()`) solo con números.
*   **`df.describe()`**: Es la forma más rápida de obtener un perfil estadístico completo. Si se usa con `exclude='number'`, se adapta para resumir categorías mostrando cuántos valores distintos existen y cuál es el "top" o moda.
