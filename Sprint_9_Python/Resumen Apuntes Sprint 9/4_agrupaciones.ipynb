```python
df = sns.load_dataset("penguins")
df.head()
```


<div style="background-color: #FADBD8; padding: 30px;">
    GROUP BY.
</div>



----

### 1 VAR CATEGORICA

----

```python
# NO IMP
# Agrupación básica por una variable categórica (especie).
# Ejecutar solo el groupby devuelve un objeto agrupador, no un resultado visual útil todavía. No util, lo util es para calcular las fcs de agregacion sum,min, count
df.groupby('species')
```

```python
# NO ASI
# Para obtener resultados, se aplican funciones de agregación como sum (suma), mean (media) o count (recuento).

# Si se aplica directamente al dataframe, se intentará aplicar a todas las columnas.

# OJO CON LA SUM PQ TE SUMA EL TEXTO COMO TIPO CONCAT Y ADEMAS RECOMIENDA PONERLO ENTRE CORCHETES

df.groupby("species").sum()
```

----

### 1 VARS NUMERICA: CALCULO DE FC AGG DE NUMERICA POR ESPECIE

----

```python
# ASI
# calculo de la media de peso por especie
# Selección de una columna específica para el resumen (devuelve una Serie). mejor asi no sino q devuleva tipo df
df.groupby("species")[["body_mass_g"]].mean()

# Uso de doble corchete para obtener el resultado en formato DataFrame.
# Hace la media del body_mass_g por cada especie
```

----

### 2 VARS NUMERICAS: AGRUPACION POR 2 VAR NUM

----

```python
# AGRUPACION POR 2 VARS NUMERICAS
# ASI SI LE PIDO 2 VARIABLES DE 2 COLUMNAS EJ LAS MEDIAS
# Agrupación por múltiples columnas numéricas (masa corporal y longitud de aleta).
# Es una agrupacion por 2 variables numericas.
df.groupby('species')[['body_mass_g', 'flipper_length_mm']].mean()
```

----

### AGRUPACION POR MAS DE 1 VAR CAT (species, sex)

----

```python
# AHORA 2 VARIABLES CATEGORICAS (SPECIES Y SEX)
# Agrupación por múltiples variables categóricas (especie y sexo).
# Se puede hacer mean, count, sum (ojo con sum), describe (resumen estadistico)
df.groupby(['species', 'sex'])[['body_mass_g', 'flipper_length_mm']].mean()
```

```python
# Obtención de un resumen estadístico completo (media, desviación, cuartiles, etc.) usando .describe().
df.groupby('species')[['body_mass_g', 'flipper_length_mm']].describe()
```

## CON 2 FCS AGG DISTINTAS , EJ MEAN Y MIN

```python
# Uso de .agg() (o .aggregate()) con un diccionario para aplicar funciones distintas a cada columna.
df.groupby(['species', 'sex'])[['body_mass_g','flipper_length_mm']].agg({
    'body_mass_g': 'mean',
    'flipper_length_mm': 'min'
})
```

```python
# Uso de .agg() (o .aggregate()) con un diccionario para aplicar funciones distintas a cada columna y sacar varias fcs de agregacion el min y el max
# Aplicación de múltiples funciones de resumen a una misma columna mediante una lista.
df.groupby(['species', 'sex'])[['body_mass_g','flipper_length_mm']].agg({
    'body_mass_g': 'mean',
    'flipper_length_mm': ['min','max']})
```


<div style="background-color: #FADBD8; padding: 30px;">
    CROSSTAB.
</div>



```python
# CROSSTAB  es una funcion no un metodo

# La función crosstab crea una matriz de frecuencias entre variables categóricas (filas species aqui y columnas sex aqui).
# si no especifico nada el crosstab hace un recuento pero puedo hacer otras agg
pd.crosstab(df['species'], df['sex'])
```

### MARGENES

```python

# Inclusión de márgenes para calcular los totales por filas y por columnas.
pd.crosstab(df['species'], df['sex'], margins=True)
```

### NORMALIZACION

```python
# Normalización de los datos para convertirlos en porcentajes sobre el total ya sea por filas como por columnas, es con el parametro normalize = all
pd.crosstab(df['species'], df['sex'], normalize='all')*100
```

```python
# Normalización por filas ('index') o por columnas ('columns'), útil para gráficos apilados.
pd.crosstab(df['species'], df['sex'], normalize='index')*100

# ESTO UTIL PARA HACER GRAFICOS APILADOS
```

```python
# Por columnas
pd.crosstab(df['species'], df['sex'], normalize='columns')*100
```

```python
# TODO
# HACER UNA FUNCION SOBRE EL BODY MASS ES CON AGGFUNC

# Uso de crosstab para calcular la media de una variable numérica (flipper_length_mm) en el cruce de categorías.
# Esto funciona de forma similar a una transposición o pivotado de datos.
pd.crosstab(
    index=df['species'],
    columns=df['sex'],
    values=df['flipper_length_mm'],
    aggfunc='mean'
)

```


**Resumen de los conceptos clave aplicados en el script:**

RESUMEN DE VAR CAT POR VAR NUM
*   **Funciones de agregación:** Se utilizan para resumir los datos (suma, media, recuento, etc.) tras agruparlos.
*   **Formato de salida:** El uso de doble corchete asegura que el resultado se visualice como un DataFrame de Pandas en lugar de una serie.
*   **Flexibilidad con `.agg()`:** Permite personalizar qué estadística se calcula para cada columna específica.
*   **Tablas Cruzadas:** `crosstab` facilita la visualización de frecuencias y permite normalizar los datos para obtener porcentajes, lo que resulta fundamental para la creación de gráficos posteriores.
