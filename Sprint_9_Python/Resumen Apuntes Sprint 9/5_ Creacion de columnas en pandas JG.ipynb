```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
# 0. Creación del DataFrame inicial
# El video menciona un dataframe con nombres, edades y salarios.
df = pd.DataFrame({
    'ID': [1,2,3,4],
    'Nombre': ['Ana', 'Juan', 'Luis', 'Marta'],
    'Edad': [25,30,22,28],
    'Salario': [3200, 1500, 2800,1200]
})

df.head()
```

---
### AÑADIR UNA COLUMNA QUE SEA UNA CTE
---

```python
# 1. Añadir una columna constante - POCO UTIL salvo para crear una columna que tenga un valor y luego hacer algo con ella
df['País'] = 'España'
df
```

---
### HACER 1 OPERACION SOBRE UNA COLUMNA
---

```python
# 2. Operación sobre una columna (Salario anual)
# Se recomienda no usar espacios en los nombres de las columnas para facilitar el acceso con punto.
df['Salario_Anual'] = df['Salario'] * 12
df
```

---
### CREAR UNA COLUMNA POR CONDICION
---

```python
# NO GUSTA MEJOR CON .LOC
# 3. Condicional simple con np.where
# Si la edad es <= 25 es 'muy joven', de lo contrario 'adulto joven'.
df['grupo_edad'] = np.where(df['Edad'] <= 25, 'muy joven', 'adulto joven')
df
```

---
### CREAR UNA COLUMNA POR CONDICION FELXIBLE USANDO .LOC
---

```python
# 4. Condicionales flexibles usando .loc
# Permite asignar valores basados en condiciones específicas, la columna de status_economico al no existir la cre, MUY UTIL

df.loc[df['Salario'] > 2000, 'Edad'] # me dara la edad de las personas con salario > 2000

# SI LE PIDO UNA COLUMNA QUE NO EXISTE ENTONCES LA CREARA POR EJ STATUS ECONOMICO Y  SI LE ASIGNO UN VALOR (RICO) Y A ESTA COLUMNA SE LE DARA EL VALOR RICO CUANDO SALARIO SEA  2000
df.loc[df['Salario'] > 2000, 'status_economico'] = 'rico'
df.loc[df['Salario'] <= 2000, 'status_economico'] = 'no tan rico'
df
```

---
### APLICAR FC A 1 COL
---

```python
# 5. Aplicar una función con lambda (Mayúsculas)
# Se usa .apply para transformar cada valor de la columna nombre a mayúsculas.
df['nombre_mayusculas'] = df['Nombre'].apply(lambda x: x.upper())

# es lo mismo que poner toda la fc
def mayuscula(x):
    return x.upper()
df
```

```python

# 6. Aplicar una función definida explícitamente (Tramos de impuestos)
def salario_neto(salario):
    if salario < 1100:
        return salario
    elif 1100 <= salario < 1800:
        return salario * 0.9  # 10% impuesto
    elif 1800 <= salario < 2500:
        return salario * 0.8  # 20% impuesto
    else:
        return salario * 0.7  # 30% impuesto

df['Salario_Neto'] = df['Salario'].apply(salario_neto)
df
```

---
### AÑADIR VALORES ALEATORIOS
---

```python
# 7. Añadir valores aleatorios
# Genera 4 números aleatorios entre 50 y 99 según la cantidad de filas.
df['Puntos'] = np.random.randint(50, 100, size=len(df))
df
```

---
### GUARDAR EL INDICE COMO UNA COLUMNA
---

```python
# 8. Guardar el índice como una columna
# Útil para mantener la trazabilidad si se reordena el dataframe.
df['indice_original'] = df.index
df
```

---
### AÑADIR UNA COLUMNA NUEVA A PARTIR DE LOS VALORES DE UNA LISTA
---

```python
# 9. Añadir columna a partir de una lista
df['Grupo'] = ['A', 'B', 'C', 'D']
df
```

---
### CREAR COLUMNA VACIA Y LLENARLA
---

```python
# 10. Crear columna vacía y llenarla
df['vacio'] = np.nan
# Llenamos los nulos con 500
df['vacio'] = df['vacio'].fillna(500, inplace = True)
df

```

```python
# si quiero que aparezca como int añadiria

df['vacio'] = df['vacio'].astype("i")
df
```

---
### CREAR UN RANKING
---

```python
# 11. Crear un ranking
# Ranking descendente basado en el salario.
df['Salario_Ranking'] = df['Salario'].rank(ascending=False).astype("i")
df

# Mostrar el resultado final
print(df)
```



### Notas adicionales basadas en las fuentes:
*   **Acceso a columnas:** El autor recomienda **no poner espacios** en los nombres de las columnas para poder acceder a ellas usando la notación de punto (ej. `df.nombre`).
*   **Uso de .loc:** Es preferible sobre `np.where` cuando se necesitan manejar múltiples condiciones de manera estructurada.
*   **Persistencia de cambios:** Al usar métodos como `fillna`, recuerda que no se guardan automáticamente a menos que asignes el resultado de nuevo a la columna o uses el parámetro `inplace=True` (aunque este último no siempre está disponible en todos los métodos o versiones).
