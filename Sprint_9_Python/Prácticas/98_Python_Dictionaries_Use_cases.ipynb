## 1. Storing User Details

```python
user_details = {"name": "Taylor", "age": 34, "city": "New York"}
print(user_details)
```

## 2. Counting Occurrences

```python
sentence = "Python is great and Python is fun"
word_count = {}
for word in sentence.split():
    word = word.lower() # Normalizar las palabras a lower
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```

## 3. Quick Access to Information

```python
print(user_details["age"])

# Tips for Using Dictionaries Effectively

age = user_details.get("age", 'Not found')
print(age)

gender = user_details.get("gender", 'Not specified')
print(gender)

# Avoid Duplicate Keys: new values to an existing key -->  update de value

user_details["age"] = 35.   # Update age value
print (user_details)

# Iterate Over Dictionaries: 

for (key, value) in user_details.items():
    print(f"{key}:{value}")
```
