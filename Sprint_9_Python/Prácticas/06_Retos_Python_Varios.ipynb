# Retos Python X : Conversor de moneda

```python
# EURO/DOLLAR/YUAN
```

```python
import pandas as pd

# Importamos dataframe con tabla conversiópn desde excel 
df = pd.read_excel('Conversor.xlsx',index_col = 0, skiprows = 1)


def conversor(divisa_inicial, divisa_final, valor, df = df):
    """
    Está función espera como calor de entrada, la divisa_inicial a convertir, 
    la divisa_final que se quiere obtener y el importe que se quiere convertir.
    """
    # truncamos divisas
    divisa_inicial = divisa_inicial.upper()
    divisa_final = divisa_final.upper()

    # retornamos resultado función operada
    return (df.loc[divisa_inicial, divisa_final] * valor)

# Solicitud de datos a operar por la función
moneda_original = input ("¿Que divisa quieres convertir (origen), EURO / DOLLAR / YUAN? : ")
moneda_destino = input ("¿Cual es la divisa que quieres tener (destino), EURO / DOLLAR / YUAN?")
cantidad = float(input("Cuanta divisa quieres convertir?").replace(",", "."))

#llamada a la función
resultado = conversor(moneda_original, moneda_destino, cantidad)

print(f"La conversioón de {cantidad} {moneda_original} a {moneda_destino} da como resultado {resultado} {moneda_destino}")
```
