```python
# tablas resumen , equivalente a tablas dinamicas  
```

```python
### (Group by -- tablas por filas)
```

```python
df.groupby(["carrier"])[["arr_delay"]].mean()  # sum / count / min / max / describe.  agregadores
```

```python
df.groupby(["carrier"])[["arr_delay"]].mean().sort_values("arr_delay", ascending=False)     # ordenar por la columna deseada
```

```python
df.groupby(["carrier", "origin"]) [["arr_delay", "dep_delay", "air_time", "distance"]].describe().to_excel("hola.xlsx", engine="openpyxl")
```

```python
### (crosstab  -- para hacer tablas de doble entrada)
```

```python
pd.crosstab(df["origin"], df["dest"])
```
