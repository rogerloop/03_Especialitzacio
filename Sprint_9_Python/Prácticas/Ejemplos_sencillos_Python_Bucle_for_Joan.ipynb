# Ejemplo 1 - Notas

```python
notas = [4,6,3,2,8,9,7,3]

s = "suspendido"
a = "aprobado"
b = "bien"
n = "notable"
e = "excelente"

for nota in notas:
    if nota < 5: 
        evaluacion = s
    elif nota < 6:
        evalucion = a
    elif nota < 7:
        evaluacion = b
    elif nota < 9:
        evaluacion = n
    else:
        evaluacion = e

    print ("Con tu nota", nota, " obtienes un: ", evaluacion)
```
