# Ejemplo buenas practicas para funciones

```python
def funcion_suma(datos):
    """
    Suma de los números de una lista o similares 
    datos: lista, tupla o set
    """

    if isinstance(datos, (list, tuple, set)):
        acum = 0
        for dato in datos:
            acum += dato

        return acum
    else:
        print("Formato erroneo")
```

```python
milista = [4,5,6,7,8,9,10]
resultado = funcion_suma(milista)
resultado
```

```python
funcion_suma((4))
```

```python
funcion_suma((a,b,c))
```
