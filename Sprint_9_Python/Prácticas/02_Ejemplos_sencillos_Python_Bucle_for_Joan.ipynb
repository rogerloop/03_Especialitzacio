# Ejemplo 1 - Notas

```python
# Bucle for or valor

notas = [4,6,3,2,8,9,7,3]

s = "suspendido"
a = "aprobado"
b = "bien"
n = "notable"
e = "excelente"

for nota in notas:
    if nota < 5: 
        evaluacion = s
    elif nota < 6:
        evalucion = a
    elif nota < 7:
        evaluacion = b
    elif nota < 9:
        evaluacion = n
    else:
        evaluacion = e

    print ("Con tu nota", nota, " obtienes un: ", evaluacion)
```

```python
# Bucle for por posición

notas = [4,6,3,2,8,9,7,3]

s = "suspendido"
a = "aprobado"
b = "bien"
n = "notable"
e = "excelente"

for i in range(len(notas)):
    if notas[i] < 5: 
        evaluacion = s
    elif notas[i] < 6:
        evalucion = a
    elif notas[i] < 7:
        evaluacion = b
    elif notas[i] < 9:
        evaluacion = n
    else:
        evaluacion = e

    print ("Con tu nota", notas[i], " obtienes un: ", evaluacion)
```

# Ejemplo 2

```python
import pandas as pd

datos = [
    ["Manzana", "Fruta", 0.99],
    ["Pollo", "Caene", 3.50],
    ["Cola", "Bebida", 1.55],
    ["Piña", "Fruta", 2.10],
    ["Leche", "Lácteos", 0.78],
    ["Pera", "Fruta", 1.58],
    ["Yogurt", "Lácteos", 0.80]
]

df = pd.DataFrame(datos, columns = ["Producto", "Tipo", "Precio"])
df.index = ["P" + str(i) for i in df.index]
df
```

```python
for fila in df.index:
    print(fila)
```

```python
# Por valor
gastofruta = 0
for fila in df.index:
    if df.loc[fila, "Tipo"] == "Fruta":
        print("Tienes un descuento del 20%, ", round(df.loc[fila, "Precio"] * 0.8, 2))
        gastofruta += round(df.loc[fila, "Precio"] * 0.8, 2)

print(round(gastofruta, 2))
```

```python
# Por posición
gastofruta = 0
for i in range(df.shape[0]):
    if df.iloc[i, 1] == "Fruta":
        print("Tienes un descuento del 20%, ", round(df.iloc[i, 2] * 0.8, 2))
        gastofruta += round(df.iloc[i, 2] * 0.8, 2)

print(round(gastofruta, 2))
```

```python
# iterar por columnas

from pandas.api.types import is_numeric_dtype
import matplotlib.pyplot as plt

for columna in df.columns:
    if is_numeric_dtype(df[columna]):
        df[columna].plot.hist()
    else:
        df[columna].value_counts().plot.pie()
    plt.show()
```
