#### El cálculo del IMC

Para calcular el IMC es necesario dividir el peso en kg de un paciente entre el cuadrado de su altura en metros. Con esto obtendremos un valor que se situará en uno de los siguientes rangos:

Fórmula IMC = Peso Kg / (Altura m)**2

- Bajo peso: menos de 18,5 
- Peso normal: 18,5-24,9 
- Sobrepeso: 25-29,9 
- Obesidad: más de 30

```python
import numpy as np
import pandas as pd
```

```python
"""
Aplicación para calcular el IMC y categorizar el estado de salud
según el índice de masa corporal.
"""
# Funciones auxiliares
# _____________________________________________________
def peticion_datos(peticion):
    """
    Solicita al usuario los datos corporales a introducir y 
    retorna la variable respuesta que será validada fuera de la función
    """
    while True:
        try:
            entrada = input(peticion)
            # Sustituir coma por punto para evitar error al convertir STR a float
            entrada = entrada.replace(",", ".")
            respuesta = float(entrada)
            
            return respuesta

        except ValueError:            
            print("Entrada no válida. Introduce un número.")
    
def calculo_imc(peso, altura):
    """Esta función calcula y retorna el IMC
    índice de masa corporal, según los argumentos 
    peso y altura    
    """
    imc = round(peso / (altura**2),1)
    return imc

def categorizar(indice):
    """
    Esta función recibe un indice IMC y categoriza
    el estado en funcion de ese IMC, retornando la 
    categoria y una receta para cada estado
    """
    if indice < 18.5:
        cat = "bajo peso"
        receta = "sigue así tu salud lo agradecerá ;)"
    elif indice <= 24.9:
        cat = "peso normal"
        receta = "lo estás haciendo bién, mantén esta línea :)"
    elif indice <= 29.9:
        cat = "sobrepeso"
        receta = "cuidado, deberías hacer más ejercicio y vigilar la alimentación."
    else:
        cat = "obesidad"
        receta = "con este indice, estás poniendo en peligro tu vida, " \
                    "\ndeberías ir al médico con urgencia para encontrar solución"
    return(cat, receta)

def obtencion_datos_validados():
    """
    No recibe parametros, llama a la función peticion_datos dos veces
    y obtiene el pes y el alto que lo retorna mediante return
    """
    peticion_pes = "Por favor, ¿Nos puedes indicar tu peso en kilogramos (solo valor)? " \
                        "\nintroduce un peso coherente entre 30 y 250 Kg: "
    peticion_alto = "¿Nos puedes facilitar tu altura en metros (solo valor)? " \
                        "\nintroduce una altura coherente entre 1 y 2.5 m: "

    while True: 
        pes = peticion_datos(peticion_pes)      
        if 30 <= pes <= 250:
            break     
        else:
            print("Valor de peso incorrecto o fuera de rango")  

    while True:
        alto = peticion_datos(peticion_alto)
        if 1 <= alto <= 2.5:
            break
        else:
            print("Valor de altura incorrecto o fuera de rango")

    return(pes, alto)

# Función orquestadora MAIN para llamar al restio de funciones
#_______________________________________________________

def main():
        
    # mensaje para usuario explica lo que hará la app
    print("Vamos a analizar tu índice de masa corporal (IMC), " \
        "\npara poder calcularlo vamos a necesitar la siguiente información:")
    
    # llamamos funcion para obtener datos usuario validados y pasarlos al calculo
    pes, alto = obtencion_datos_validados()

    # pasamos los datos del usuario a la duncion calculo para obtener IMC
    calculo = calculo_imc(pes, alto)

    # pasamos a funcion categorizar el calculo IMC para que categorice
    categoria, propuesta = categorizar(calculo)

    # Respuesta al usuario del analisis y categorizacion de IMC
    print(f"Despúes de analizar tu peso de {pes} Kg. y altura de {alto} m, podemos afirmar que: \
        \nTu IMC es {calculo}, esto indica que estas en la categoría de {categoria},\
            \nnuestro consejo en funión de tu IMC es: {propuesta} ")

if __name__ == "__main__":
    main()
```

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

```python
from IPython.display import clear_output

# para toda la app para palabra escala usamos diminutivo (esc)

def mostrar_menu(cabecera, texto, opcion):
    ''' 
    Funcion que genera el menu personaliado en funcion de si se la llama 
    para escalas iniciales o para escalas finales, retorna el elemento del menú seleccionado
    '''
    salir = False
    
    print("\n", cabecera, "\n")
   
    while not salir:
        try:

            print(f'''Escala:
            1. Celsius (C)
                    
            2. Fahrenheit (F)
                    
            3. Kelvin (K)
                    
            4. Rankine (Ra)
                    
            5. Réamur (Re)   
                        
            {opcion}''')
        
            clear_output(wait=True)
    
            return int(input(texto))
         
        except Exception as e:
            print(f'Ocurrio un error: {e}')


def esc_entrada():
    ''' 
    No recibe parametros se usa para llamar al menu de seleccion y 
    devuelve la escala de temperaturas del valor inicial
    '''
  
    titulo_menu = "** Escalas de temperatura de entrada para convertir **"
    texto = "Elige la escala de temperatura inicial (1-5) o (9) para salir de la app: "
    opcion_salir = "9. Salir para finalizar Aplicación"
    
    
    esc_inicial = mostrar_menu(titulo_menu, texto, opcion_salir)
    while esc_inicial not in (1,2,3,4,5,6,9):
        esc_inicial = mostrar_menu(titulo_menu, texto, opcion_salir)
        print('*** Atención! Introduce un número válido***')

    return esc_inicial


def peticion_valor():
    """
    Solicita al usuario el valor de temperatura que se quiere convertir  
    retorna la variable valor que será validada fuera de la función
    """
    while True:
        try:            
            entrada = input("Introduce el valor a convertir: ")
            # Sustituir coma por punto para evitar error al convertir STR a float
            entrada = entrada.replace(",", ".")
            valor = float(entrada)
                
            return valor

        except ValueError:            
            print("Entrada no válida. Introduce un número valido.")


def validacion_valor(escala):
    """
    Recibe como parametro, la escala sleccionada de entrada (escala)
    llama a la función peticion_valor y obtiene el valor de temperatura que se quiere convertir
    que despues de validar si esta en rango se pasa mediante return
    Para validar llama la funcion conversor_to_c para comprobar si temperatura en celsius es nenor que -273.15 celsius
    """
 
    while True: 
        temp = peticion_valor()
        temperatura_celsius = coversor_to_c(temp, escala)

        if temperatura_celsius < -273.15:
            print("ATENCIÓN!! Valor de temperatura fuera de rango para la escala seleccionada: \n" \
                    "Min (C): -273.15, Min (K): 0, Min (F): -459.67, Min (Ra): 0, Min (Re): -218.52")               
        else:
            break              

    return temperatura_celsius
    

def esc_salida():
    ''' 
    Esta funcion no recibe parametros y sirve para personalizar las opciones de la funcion mostrar menu
    cada vez que se la llama  devuelve mediante return la escala seleccionada cada vez
    '''
    titulo_menu = "\n** Escalas de temperatura de salida a las que quieres convertir **"
    texto = "Elige las escalas de temperatura final que deseees (1-5) \n" \
    "       elige (6) para finalizar la selección: "
    opcion_fin = "6. Finalizar la selección de escalas de conversión \n"
    esc_final= mostrar_menu(titulo_menu, texto, opcion_fin)
    return esc_final


def bucle_esc_salida():
    ''' 
    Esta función no recibe parametros 
    llena la lista de tipo de escala de salida mediante un bucle, llama la funcion esc_salida
    que a su vez usa la funcion mostrar menu para retornar la lista de escalas a las que se desea convertir
    '''
    lista_esc_salida = []
    while True:
        escalas_salida = esc_salida()
        if escalas_salida == 6:
            break
        else:
            lista_esc_salida.append(escalas_salida)

    return lista_esc_salida


def coversor_to_c(temperatura, esc_ini):
    ''' 
    Esta función recibe por parametros el valor a convertir (temperatura) y la escala que es.
    Convierte cualquier valor a Celsius y lo retorna.
    '''

    if esc_ini == 1:
        celsius = temperatura                   # C Celsius to C Celsius

    elif esc_ini == 2:
        celsius = (temperatura-32)*5/9          # F Fahrenheit to C Celsius

    elif esc_ini == 3:
        celsius = temperatura - 273.15          # K Kelvin to C Celsius

    elif esc_ini == 4:
        celsius = (temperatura-491.67)*5/9      # Ra Rankine to C Celsius

    else:
        celsius = temperatura*5/4               # Re Réaumur to C Celsius

    return celsius


def conversor_from_c(temp_cel, esc_fin):
    ''' 
    Recibe por parametros la temperatura en Cenlsius y la lista con las escalas a las que se quiere convertir
    crea un diccionario que rellena con las conversiones de las escalas finales seleccionadas
    '''
    conversion = {}

    for escala in esc_fin:

        if escala == 1:
            temp_convertida = temp_cel                  # C Celsius to C Celsius
            conversion['Celsius'] = temp_convertida

        elif escala == 2:
            temp_convertida = (temp_cel*9/5)+32         # C celsius to F Fahrenheit
            conversion['Fahrenheit'] = temp_convertida

        elif escala == 3:
            temp_convertida = temp_cel + 273.15         # C Celsius to K Kelvin
            conversion['Kelvin'] = temp_convertida

        elif escala == 4:
            temp_convertida = (temp_cel+273.15)*9/5     # C Celsius to Ra Rankine
            conversion['Rankine'] = temp_convertida

        else:
            temp_convertida = temp_cel*4/5              # C Celsius to Re Réaumur
            conversion['Réaumur'] = temp_convertida

    return conversion
 
        
def mostrar_resultat(temperatura_final):
    ''' 
    Esta función recibe el diccionario conversion como argumentto (temperatura_final), 
    itera el diccionario y muestra meduiante print las temperaturas convertidas
    '''

    print(f"\nLa temperatura convertida es: ") 
    
    for escala, valor in temperatura_final.items():

        print(f"{escala}: {valor}")
    
    print("\nSi quieres salir de la Aplicación selecciona (9 + intro)")
    
def main():
    ''' 
    Es la función orquestadora que organiza el resto de funciones y
    el flujo de la App. 
    '''
    while True:                             # Mantiene la App activa hasta que usuario introduzca 9
        escala_ini = esc_entrada()          # Llamada a la función donde se selecciona la escala de entrada
        if escala_ini == 9:
            print("Saliendo de la App")
            break
        
        # Llamada a la función que valida el valor de temperatura a convertir
        # Desde dentro de esta función se llama a la función peticion_valor y la funcion conversor to C
        temperatura_cel = validacion_valor(escala_ini)      

        # Llamada a la funcion que genera un bucle para poder seleccionar varias escalas de salida
        escala_fin = bucle_esc_salida()

        # Llamada a la funcion que realiza la conversión de Celsius a las escalas de salida seleccionadas
        temp_fin = conversor_from_c(temperatura_cel, escala_fin)

        # Llamada a la función que muestra los resultados de la conversión
        mostrar_resultat(temp_fin)       

 
if __name__ == "__main__":
    main()
```

#### 3. Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

```python

def demanem_text():
    '''
    Sol·licita a l'usuari la introducció d'un text per consola.
    Controla que l'entrada no sigui buida ni contingui només espais en blanc.
    Manté un bucle fins que l'usuari introdueix un text vàlid.

    Returns:
        str: Text introduït per l'usuari, net d'espais als extrems.
    '''
    # Uso metodo .strip() para eliminar los espacios en blanco delante y detras de text
    while True:
        text = input("Si us plau introdueix qualsevol text: ").strip()
        
        if text:
            return text
        else:
            print("Error: has d'introduir algun text, no pot estar buit.")
            

def netejar_frase(frase):
    '''
    Rep una frase i la neteja eliminant caràcters no alfabètics.
    Separa la frase en paraules i retorna una llista amb les paraules netes.

    Args:
        frase (str): Text d'entrada a processar.

    Returns:
        list: Llista de paraules només amb caràcters alfabètics.
    '''
    paraules = frase.split()
    paraules_netes = []

    for paraula in paraules:   
        paraula_neta = ""     
        for caracter in paraula:
            if caracter.isalpha():
                paraula_neta += caracter

        if paraula_neta:
            paraules_netes.append(paraula_neta)
    
    return paraules_netes


def paraules_count(paraules):
    '''
    Rep una llista de paraules i compta quantes vegades apareix cadascuna.

    Args:
        paraules (list): Llista de paraules netes.

    Returns:
        dict: Diccionari amb les paraules com a claus i el nombre d'aparicions com a valors.
    '''
    diccionari = {}

    for paraula in paraules:
        if paraula in diccionari:
            times = diccionari.get(paraula)
            times += 1
            diccionari[paraula] = times
        else: 
            diccionari[paraula] = 1

    return diccionari


def llargada_mitjana_paraules(paraules):
    '''
    Calcula la llargada mitjana de les paraules d'una llista.

    Args:
        paraules (list): Llista de paraules netes.

    Returns:
        float: Mitjana de caràcters per paraula (arrodonida a 2 decimals).
    '''
    suma_caracters = 0

    if len(paraules) > 0:
        for paraula in paraules:
            suma_caracters += len(paraula)

        mitjana = round(suma_caracters / len(paraules), 2)
    else:
        mitjana = 0

    return mitjana


def mostrar_resultat(diccionari_paraules):
    ''' 
    Mostra per pantalla el recompte de paraules.

    Rep un diccionari amb paraula → nombre d'aparicions i imprimeix
    cada element de forma clara.

    Args:
        diccionari_paraules (dict): Diccionari amb el recompte de paraules.
    '''
    print(f"\nEl recompte de paraules del text és: \n") 
    
    for paraula, valor in diccionari_paraules.items():
        print(f"{paraula}: {valor}")


def main():
    ''' 
    Funció principal orquestradora que coordina el flux de l'aplicació.

    Passos:
    1. Demana un text a l'usuari.
    2. Neteja la frase eliminant caràcters no alfabètics.
    3. Comptabilitza les paraules.
    4. Calcula la llargada mitjana de les paraules.
    5. Mostra els resultats per pantalla.
    '''
    frases = demanem_text()
        
    frase_neta = netejar_frase(frases.lower())

    comptar_paraules = paraules_count(frase_neta)

    mitjana_paraules = llargada_mitjana_paraules(frase_neta)

    print(f"\nAnàlisi del text proporcionat: {frases}")

    mostrar_resultat(comptar_paraules)

    print(f"\nLa llargada mitjana de les paraules del text és: {mitjana_paraules} lletres per paraula")


if __name__ == "__main__":
    main()
```

#### 4. Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

```python

def revertir_dic(dic_inicial):
    ''' 
    Recibe un diccionario y invierte las (clave,valor) convirtiendo los valores
    a claves y las claves a valores del diccionario retornado.
    Además si hay valores duplicados en el diccionario de entrada, los acumula en una lista 
    bajo la misma clave y imprime un mensaje de advertencia

    Args: 
        dic_inicial (diccionario): diccionario de entrada a procesar

    Returns:
        Diccionario claves = valores entrada si duplicados se presentan 
        con valores añadidos a lista | valores = lista de claves entrada
    '''
    dic_invertido = {}

    # bucle for para iterar el diccionario mediante método .items() que me devuelve tuplas "parejas" (clave,valor)

    for clave, valor in dic_inicial.items(): 

        # comprobar si valor ya existe en dic y sino crear la nueva clave

        if valor not in dic_invertido:
            dic_invertido[valor] = [clave]

        # si existe el valor, se agrega (clave) como valor en la lista para esa "clave" que viene del valor
        # quedando esa clave como una lista convarios valores
        # impresion de claves que tienen varios valores

        else:
            dic_invertido[valor].append(clave)
            print(f"Estos valores estan duplicados '-{valor}-'")

    return dic_invertido


dictionary_inicial = {
    'a': 'apple',
    'b': 'banana',
    'c': 'cherry',
    'd': 'apple',
    'e': 'banana',
    'f': 'date',
    'g': 'fig',
    'h': 'grape',
    'i': 'fig',
    'j': 'apple'
}

# Llamada a la funcion para revertir clave, valor del diccionario facilitado

diccionario_invertido = revertir_dic(dictionary_inicial)

#Imprimimos el diccionario_invertido resultante de la salida de la función

print(diccionario_invertido)
```

```python


# Lo pongo comentario por si quiero usar despues para autoimatizar y comprobar carga archivo

""" def demanem_text(texto):
    '''
    Sol·licita a l'usuari la introducció d'un text per consola.
    Controla que l'entrada no sigui buida ni contingui només espais en blanc.
    Manté un bucle fins que l'usuari introdueix un text vàlid.

    Returns:
        str: Text introduït per l'usuari, net d'espais als extrems.
    '''
    while True:
        text = texto.strip()
        
        if text:
            return text
        else:
            print("Error: has d'introduir algun text, no pot estar buit.") """
            

import unicodedata  # cargo unicode para poder eliminar  acentos de las palabras

def eliminar_acentos(texto):
    """ 
    Recibe texto y elimina todos acentos de las palabras

    Args:
        texto (str): Texto de entrada a procesar

    Returns: Retorna el mismo texto pero sin acentos
    """
    texto = unicodedata.normalize('NFD', texto)
    texto = ''.join(c for c in texto if unicodedata.category(c) != 'Mn')
    return texto

def netejar_frase(frase):
    '''
    Rep una frase i la neteja eliminant caràcters no alfabètics.
    Separa la frase en paraules i retorna una llista amb les paraules netes.

    Args:
        frase (str): Text d'entrada a processar.

    Returns:
        list: Llista de paraules només amb caràcters alfabètics.
    '''
    paraules = frase.split()
    paraules_netes = []

    for paraula in paraules:   
        paraula_neta = ""     
        for caracter in paraula:
            if caracter.isalpha():
                paraula_neta += caracter

        if paraula_neta:
            paraules_netes.append(paraula_neta)
    
    return paraules_netes


def paraules_count(paraules):
    '''
    Rep una llista de paraules i compta quantes vegades apareix cadascuna.

    Args:
        paraules (list): Llista de paraules netes.

    Returns:
        dict: Diccionari amb les paraules com a claus i el nombre d'aparicions com a valors.
    '''
    diccionari = {}

    # Bucle [clave] valor para añadir nº veces aparece palabra al diccionario

    for paraula in paraules:
        if paraula in diccionari:
            times = diccionari.get(paraula)
            times += 1
            diccionari[paraula] = times
        else: 
            diccionari[paraula] = 1

    return diccionari


def agrupar_letra(diccionario_palabras):
    '''
    Agrupa un diccionario de palabras por su letra inicial (a-z).

    Args:
        diccionario_palabras (dict): {'palabra': conteo}

    Returns:
        dict: {'a': {'agua': 2, ...}, 'b': {...}, ...}
    '''
    
    diccionario_agrupado = {'a': {}, 'b': {}, 'c': {}, 'd': {}, 'e': {}, 'f': {},
                             'g': {}, 'h': {}, 'i': {}, 'j': {}, 'k': {}, 'l': {}, 'm': {},
                               'n': {}, 'ñ': {}, 'o': {}, 'p': {}, 'q': {}, 'r': {}, 's': {},
                                 't': {}, 'u': {}, 'v': {}, 'w': {}, 'x': {}, 'y': {}, 'z': {}}

    # Omplir diccionari

    for palabra, conteo in diccionario_palabras.items():

        inicial = palabra[0]

        diccionario_agrupado[inicial][palabra] = conteo

    # Ordenar paraules dins de cada lletra

    for lletra in diccionario_agrupado:

        diccionario_agrupado[lletra] = dict(

            sorted(diccionario_agrupado[lletra].items())

        )

    return diccionario_agrupado


def mostrar_resultat(diccionari_ordenat):
    ''' 
    Mostra per pantalla el recompte de paraules.

    Rep un diccionari amb paraula → nombre d'aparicions i imprimeix
    cada element de forma clara.

    Args:
        diccionari_paraules (dict): Diccionari amb el recompte de paraules.
    '''
    print(f"\nEl recompte de paraules del text és: ") 
    
    for letra, palabras in diccionari_ordenat.items():
        print(f"\n{letra}:")
        for palabra, conteo in palabras.items():
            print(f"  {palabra}: {conteo}")


def main():
    '''
    Funció principal de l'aplicació.
    Coordina tot el flux del programa:

    1. Carrega el contingut d'un fitxer de text. (with open)
    2. Converteix tot el text a minúscules.
    3. Elimina accents i caràcters especials.
    4. Neteja el text deixant només paraules alfabètiques.
    5. Compta el nombre d'aparicions de cada paraula.
    6. Agrupa les paraules segons la seva lletra inicial.
    7. Mostra el resultat final ordenat per lletres.

    Returns:
        None
    '''
    # Carga del texto a analizar
    with open ('/Users/rogerdefez/Documents/Cursos i Llibres/BootCamp IT Academy/03_Especialitzacio/Sprint_9_Python/Sprint_9_Task_S9_01/tu me quieres blanca.txt') as text:

        texto = text.read()


    texto = texto.lower()       # Convierto texto a minúsculas

    texto = eliminar_acentos(texto)

    texto_limpio = netejar_frase(texto)  

    contar_palabras = paraules_count(texto_limpio)

    diccionario_final = agrupar_letra(contar_palabras)

    print(mostrar_resultat(diccionario_final))


if __name__ == "__main__":
    main()
```

#### 2. Conversió de tipus de dades.

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:

[ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python

def separar_datos(lista_datos):
    '''
    Separa los elementos convertibles a float de aquellos
    que no pueden convertirse.

    La función recorre una lista que puede contener:
    - strings
    - números
    - listas
    - tuplas

    Si un elemento puede convertirse a float, se almacena
    en la lista de números decimales.

    En caso contrario, se almacena en la lista de datos
    no convertibles.

    Args:
        lista_datos (list): Lista con diferentes tipos de datos.

    Returns:
        tuple:
            - lista_float (list): Elementos convertidos a float.
            - lista_resto (list): Elementos no convertibles a float.
    '''
    lista_float = []
    lista_resto =[]
    for dato in lista_datos:
        
        if type(dato) is not str:
            for element in dato:
                try:
                    element = float(element)
                    lista_float.append(element)

                except:
                    lista_resto.append(element)

        else: 
            try:
                dato = float(dato)
                lista_float.append(dato)

            except:
                lista_resto.append(dato)

    return lista_float, lista_resto

def main():
    '''
    Función principal de la aplicación.

    La función:
    - Crea una lista con diferentes tipos de datos.
    - Llama a la función separar_datos().
    - Muestra por pantalla:
        - los datos convertidos a float
        - los datos no convertibles

    Returns:
        None
    '''
    lista_datos = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

    floats, resto_datos = separar_datos(lista_datos)

    print(floats)
    print(resto_datos)

if __name__ == "__main__":
    main()
```

#### 1. Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

longitud (int): Longitud de la contrasenya
majuscules (bool = True): Si hi ha d’aparèixer majúscules
minuscules (bool = True): Si hi ha d’aparèixer minúscules
numeros (bool = True): Si hi ha d’aparèixer números
signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)
Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
import numpy as np
import random
import string
import pyperclip


def main_aleatori(long, may, min, num, sim):
    '''
    Genera un array aleatorio de números enteros entre 1 y 4
    que representa los diferentes tipos de caracteres de la contraseña.

    La función valida que el array generado cumpla los requisitos
    mínimos indicados mediante los parámetros booleanos:

    1 → Mayúsculas
    2 → Minúsculas
    3 → Números
    4 → Símbolos especiales

    Mientras no se cumplan todos los requisitos obligatorios,
    seguirá generando nuevos arrays aleatorios.

    Args:
        long (int): Longitud total de la contraseña.
        may (bool): Indica si debe contener mayúsculas.
        min (bool): Indica si debe contener minúsculas.
        num (bool): Indica si debe contener números.
        sim (bool): Indica si debe contener símbolos especiales.

    Returns:
        numpy.ndarray: Array con la estructura aleatoria validada.
    '''
    
    while True:
        
        num_aleatorio = np.random.randint(1,5, size=long)

        if may and 1 not in num_aleatorio:
            continue
        if min and 2 not in num_aleatorio:
            continue
        if num and 3 not in num_aleatorio:
            continue
        if sim and 4 not in num_aleatorio:
            continue
            
        return num_aleatorio


def mayusculas():
    '''
    Genera una letra mayúscula aleatoria.

    Utiliza el módulo string para acceder al abecedario
    en mayúsculas y random.choice() para seleccionar
    un carácter aleatorio.

    Returns:
        str: Letra mayúscula aleatoria.
    '''

    letra_mayus = random.choice(string.ascii_uppercase)

    return letra_mayus


def minusculas():
    '''
    Genera una letra minúscula aleatoria.

    Utiliza el módulo string para acceder al abecedario
    en minúsculas y random.choice() para seleccionar
    un carácter aleatorio.

    Returns:
        str: Letra minúscula aleatoria.
    '''

    letra_minus= random.choice(string.ascii_lowercase)

    return letra_minus


def numeros():
    '''
    Genera un número aleatorio entre 0 y 9.

    El número generado se convierte a string para poder
    concatenarlo posteriormente a la contraseña final.

    Returns:
        str: Número aleatorio convertido a texto.
    '''

    numero = random.randint(0, 10)

    return str(numero)


def simbolos():
    '''
    Genera un símbolo especial aleatorio.

    Utiliza string.punctuation para acceder a una colección
    de caracteres especiales estándar del teclado.

    Returns:
        str: Símbolo especial aleatorio.
    '''
    
    simbolo = random.choice(string.punctuation)

    return simbolo


def crear_contrasenya(long):
    '''
    Genera una contraseña aleatoria segura.

    La función solicita previamente una estructura aleatoria
    validada mediante main_aleatori() y, en función de cada
    valor numérico recibido, añade un tipo concreto de carácter:

    1 → Mayúscula
    2 → Minúscula
    3 → Número
    4 → Símbolo especial

    Finalmente devuelve la contraseña completa generada.

    Args:
        long (int): Longitud total de la contraseña.

    Returns:
        str: Contraseña aleatoria generada.
    '''

    contrasenya = ""
    
    aleatoria = main_aleatori(long, True, True, True, True)

    for num in aleatoria:

        if num == 1:
            contrasenya += mayusculas()

        elif num == 2:
            contrasenya += minusculas()

        elif num == 3:
            contrasenya += numeros()

        else:
            contrasenya += simbolos()

    return contrasenya


def copy_portapapeles(psswrd):
    '''
    Copia automáticamente la contraseña generada
    al portapapeles del sistema operativo.

    Utiliza la librería pyperclip para permitir
    posteriormente pegar la contraseña mediante
    Ctrl + V o Cmd + V.

    Args:
        psswrd (str): Contraseña que se desea copiar.

    Returns:
        None
    '''

    pyperclip.copy(psswrd)
    print("Contraseña copiada")


def main():
    '''
    Función principal de la aplicación, la funcion orquestadora.

    Coordina todo el flujo de ejecución del generador
    de contraseñas:

    1. Solicita al usuario la longitud deseada.
    2. Controla errores mediante try/except:
       - Valores no numéricos.
       - Longitudes inferiores a 4 caracteres.
    3. Genera una contraseña aleatoria segura.
    4. Muestra la contraseña por pantalla.
    5. Permite opcionalmente copiarla al portapapeles.
    6. Controla que la respuesta del usuario sea únicamente S o N.

    Returns:
        None
    '''

    while True:

        try:
            
            long_contrasenya = int(input("Longitud de la contrasenya min. 4 caracteres? :"))

            if long_contrasenya < 4:
                print("Atención: la contraseña debe tener mínimo 4 caracteres")
                
                continue

            break

        except ValueError:

            print("Atención: introduce solo números enteros mayores de 3")


    contrasenya_creada = crear_contrasenya(long_contrasenya)

    print(f"\nEsta es la contraseña generada {contrasenya_creada}")

    while True:

        copy = input("¿Quieres que copie la contraseña en portapapeles para usar más tarde?: S/N")

        if copy.lower()  == "s":
            copy_portapapeles(contrasenya_creada)
            break

        elif copy.lower() == "n":            
            print("Fin de la App")
            break

        else:
            print("Error: introduce solo S o N")


if __name__ == '__main__':
    main()
```

#### 2. Processament de dades simple

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:

El nombre total de gols que ha fet cada equip.
El nom de l’equip més golejador.
El nom de l’equip més golejat
La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)


```python

# imprtar archivo con resultados
def validar_equipos(equipos,equipo):
    '''
    Valida si un equipo existe dentro del diccionario principal.
    Si el equipo no existe, lo inicializa creando un subdiccionario
    con todas sus estadísticas a cero.

    Args:
        equipos (dict): Diccionario principal con todos los equipos.
        equipo (str): Nombre del equipo a validar.

    Returns:
        None
    '''
    if equipo not in equipos:

        equipos[equipo]={
            "ganados": 0,
            "perdidos": 0,
            "empatados": 0,
            "goles_favor": 0,
            "goles_contra": 0,
            "puntos": 0
        }


def partidos(equipos, local, visitante , goles_local, goles_visitante):
    '''
    Procesa un partido y actualiza las estadísticas de ambos equipos.
    La función:
    - Valida que los equipos existan.
    - Actualiza goles a favor y en contra.
    - Actualiza victorias, derrotas, empates y puntos.

    Args:
        equipos (dict): Diccionario principal de equipos.
        local (str): Nombre del equipo local.
        visitante (str): Nombre del equipo visitante.
        goles_local (int): Goles marcados por el equipo local.
        goles_visitante (int): Goles marcados por el equipo visitante.

    Returns:
        None
    '''
    # validar equipos
    validar_equipos(equipos, local)
    validar_equipos(equipos, visitante)

    # actualizar goles a favor
    equipos[local]['goles_favor'] += goles_local
    equipos[visitante]['goles_favor'] += goles_visitante

    # actualizar goles en contra
    equipos[local]['goles_contra'] += goles_visitante
    equipos[visitante]['goles_contra'] += goles_local

    # evaluar puntos por resultado
    if goles_local > goles_visitante:
         equipos[local]['ganados'] += 1
         equipos[local]['puntos'] += 3

         equipos[visitante]['perdidos'] += 1

    elif goles_local < goles_visitante:
         equipos[visitante]['ganados'] += 1
         equipos[visitante]['puntos'] += 3

         equipos[local]['perdidos'] += 1

    # seria el caso empate goles_local == goles_visitante
    else: 
         equipos[local]['empatados'] += 1
         equipos[local]['puntos'] += 1
         
         equipos[visitante]['empatados'] += 1
         equipos[visitante]['puntos'] += 1

def obtener_puntos(equipo):
    '''
    Obtiene los puntos de un equipo para usar como criterio
    de ordenación en la clasificación.

    Args:
        equipo (tuple): Tupla generada por .items()
                        con formato:
                        (nombre_equipo, datos_equipo)

    Returns:
        int: Número de puntos del equipo.
    '''
    return equipo[1]['puntos']

def mostrar_clasificacion(equipos):
    '''
    Muestra la clasificación general de la competición
    ordenada de mayor a menor puntuación.

    También muestra:
    - partidos ganados
    - partidos perdidos
    - partidos empatados
    - goles a favor
    - goles en contra

    Args:
        equipos (dict): Diccionario principal con estadísticas
                        de todos los equipos.

    Returns:
        None
    '''
    print("\nCLASIFICACIÓN RESULTADOS Y GOLES DE LA COMPETICIÓN ES:\n")

    clasificacion_puntos = sorted(equipos.items(), key=obtener_puntos, reverse=True)

    for equipo, datos_equipo in clasificacion_puntos: 
        print(
                f"Puntos: {datos_equipo['puntos']}"
                f" | Equipo: {equipo}"
                f" | Partidos - ganados: {datos_equipo['ganados']}"
                f" | perdidos: {datos_equipo['perdidos']}"
                f" | empatados: {datos_equipo['empatados']}"
                f" | Goles a favor: {datos_equipo['goles_favor']}"
                f" | Goles en contra: {datos_equipo['goles_contra']}")
        

def equipo_goleador(equipos):
    '''
    Calcula y muestra el equipo con mayor número
    de goles marcados en la competición.

    Args:
        equipos (dict): Diccionario principal con estadísticas
                        de todos los equipos.

    Returns:
        None
    '''
    max_goleador = max(equipos, key=lambda equipo: equipos[equipo]["goles_favor"])
    goles_marcados = equipos[max_goleador]["goles_favor"]

    print(f"\nEQUIPO MÁS GOLEADOR:\n"
          f"{max_goleador} con {goles_marcados} goles marcados.")

        
def equipo_goleado(equipos):
    '''
    Calcula y muestra el equipo con mayor número
    de goles encajados en la competición.

    Args:
        equipos (dict): Diccionario principal con estadísticas
                        de todos los equipos.

    Returns:
        None
    '''
    max_goleado = max(equipos, key=lambda equipo: equipos[equipo]["goles_contra"])
    goles_recibidos = equipos[max_goleado]["goles_contra"]

    print(f"\nEQUIPO MÁS GOLEADO:\n"
          f"{max_goleado} con {goles_recibidos} goles encajados.")


def main():
    '''
    Función principal de la aplicación.

    La función:
    - Inicializa el diccionario principal de equipos.
    - Lee el archivo de resultados de fútbol.
    - Procesa cada línea del archivo.
    - Separa equipos y resultados.
    - Convierte los goles a valores enteros.
    - Actualiza las estadísticas de cada partido.
    - Muestra la clasificación final.
    - Muestra el equipo más goleador.
    - Muestra el equipo más goleado.

    Returns:
        None
    '''

    equipos = {}

    with open("historic partits.txt", encoding='utf-8') as resultados:

        # recorro lineas de archivo
        for partido in resultados:   
            
            # limpio \n y crea una lista partido separado por comas
            partido = partido.strip()   

            # dentro de cada partido separado por comas asigno detalle (separados por tab) a su variable 
            local, resultado, visitante = partido.split('\t') 

            # separo los coles local y visitante de cada partido
            goles_local, goles_visitante = resultado.split("-")

            goles_local = int(goles_local)
            goles_visitante = int(goles_visitante)

            # creacion diccionario de partidos
            partidos(equipos, local, visitante , goles_local, goles_visitante)
                     

    mostrar_clasificacion(equipos)


    equipo_goleador(equipos)


    equipo_goleado(equipos)


if __name__ == '__main__':
    main()
```
