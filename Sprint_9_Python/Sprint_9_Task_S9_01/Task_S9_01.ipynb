#### El cálculo del IMC

Para calcular el IMC es necesario dividir el peso en kg de un paciente entre el cuadrado de su altura en metros. Con esto obtendremos un valor que se situará en uno de los siguientes rangos:

Fórmula IMC = Peso Kg / (Altura m)**2

- Bajo peso: menos de 18,5 
- Peso normal: 18,5-24,9 
- Sobrepeso: 25-29,9 
- Obesidad: más de 30

```python
import numpy as np
import pandas as pd
```

```python
"""
Aplicación para calcular el IMC y categorizar el estado de salud
según el índice de masa corporal.
"""
# Funciones auxiliares
# _____________________________________________________
def peticion_datos(peticion):
    """
    Solicita al usuario los datos corporales a introducir y 
    retorna la variable respuesta que será validada fuera de la función
    """
    while True:
        try:
            entrada = input(peticion)
            # Sustituir coma por punto para evitar error al convertir STR a float
            entrada = entrada.replace(",", ".")
            respuesta = float(entrada)
            
            return respuesta

        except ValueError:            
            print("Entrada no válida. Introduce un número.")
    
def calculo_imc(peso, altura):
    """Esta función calcula y retorna el IMC
    índice de masa corporal, según los argumentos 
    peso y altura    
    """
    imc = round(peso / (altura**2),1)
    return imc

def categorizar(indice):
    """
    Esta función recibe un indice IMC y categoriza
    el estado en funcion de ese IMC, retornando la 
    categoria y una receta para cada estado
    """
    if indice < 18.5:
        cat = "bajo peso"
        receta = "sigue así tu salud lo agradecerá ;)"
    elif indice <= 24.9:
        cat = "peso normal"
        receta = "lo estás haciendo bién, mantén esta línea :)"
    elif indice <= 29.9:
        cat = "sobrepeso"
        receta = "cuidado, deberías hacer más ejercicio y vigilar la alimentación."
    else:
        cat = "obesidad"
        receta = "con este indice, estás poniendo en peligro tu vida, " \
                    "\ndeberías ir al médico con urgencia para encontrar solución"
    return(cat, receta)

def obtencion_datos_validados():
    """
    No recibe parametros, llama a la función peticion_datos dos veces
    y obtiene el pes y el alto que lo retorna mediante return
    """
    peticion_pes = "Por favor, ¿Nos puedes indicar tu peso en kilogramos (solo valor)? " \
                        "\nintroduce un peso coherente entre 30 y 250 Kg: "
    peticion_alto = "¿Nos puedes facilitar tu altura en metros (solo valor)? " \
                        "\nintroduce una altura coherente entre 1 y 2.5 m: "

    while True: 
        pes = peticion_datos(peticion_pes)      
        if 30 <= pes <= 250:
            break     
        else:
            print("Valor de peso incorrecto o fuera de rango")  

    while True:
        alto = peticion_datos(peticion_alto)
        if 1 <= alto <= 2.5:
            break
        else:
            print("Valor de altura incorrecto o fuera de rango")

    return(pes, alto)

# Función orquestadora MAIN para llamar al restio de funciones
#_______________________________________________________

def main():
        
    # mensaje para usuario explica lo que hará la app
    print("Vamos a analizar tu índice de masa corporal (IMC), " \
        "\npara poder calcularlo vamos a necesitar la siguiente información:")
    
    # llamamos funcion para obtener datos usuario validados y pasarlos al calculo
    pes, alto = obtencion_datos_validados()

    # pasamos los datos del usuario a la duncion calculo para obtener IMC
    calculo = calculo_imc(pes, alto)

    # pasamos a funcion categorizar el calculo IMC para que categorice
    categoria, propuesta = categorizar(calculo)

    # Respuesta al usuario del analisis y categorizacion de IMC
    print(f"Despúes de analizar tu peso de {pes} Kg. y altura de {alto} m, podemos afirmar que: \
        \nTu IMC es {calculo}, esto indica que estas en la categoría de {categoria},\
            \nnuestro consejo en funión de tu IMC es: {propuesta} ")

if __name__ == "__main__":
    main()
```

#### 2. Convertidor de temperatures

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

```python
from IPython.display import clear_output

# para toda la app para palabra escala usamos diminutivo (esc)

def mostrar_menu(cabecera, texto, opcion):
    ''' 
    Funcion que genera el menu personaliado en funcion de si se la llama 
    para escalas iniciales o para escalas finales, retorna el elemento del menú seleccionado
    '''
    salir = False
    
    print("\n", cabecera, "\n")
   
    while not salir:
        try:

            print(f'''Escala:
            1. Celsius (C)
                    
            2. Fahrenheit (F)
                    
            3. Kelvin (K)
                    
            4. Rankine (Ra)
                    
            5. Réamur (Re)   
                        
            {opcion}''')
        
            clear_output(wait=True)
    
            return int(input(texto))
         
        except Exception as e:
            print(f'Ocurrio un error: {e}')


def esc_entrada():
    ''' 
    No recibe parametros se usa para llamar al menu de seleccion y 
    devuelve la escala de temperaturas del valor inicial
    '''
  
    titulo_menu = "** Escalas de temperatura de entrada para convertir **"
    texto = "Elige la escala de temperatura inicial (1-5) o (9) para salir de la app: "
    opcion_salir = "9. Salir para finalizar Aplicación"
    
    
    esc_inicial = mostrar_menu(titulo_menu, texto, opcion_salir)
    while esc_inicial not in (1,2,3,4,5,6,9):
        esc_inicial = mostrar_menu(titulo_menu, texto, opcion_salir)
        print('*** Atención! Introduce un número válido***')

    return esc_inicial


def peticion_valor():
    """
    Solicita al usuario el valor de temperatura que se quiere convertir  
    retorna la variable valor que será validada fuera de la función
    """
    while True:
        try:            
            entrada = input("Introduce el valor a convertir: ")
            # Sustituir coma por punto para evitar error al convertir STR a float
            entrada = entrada.replace(",", ".")
            valor = float(entrada)
                
            return valor

        except ValueError:            
            print("Entrada no válida. Introduce un número valido.")


def validacion_valor(escala):
    """
    Recibe como parametro, la escala sleccionada de entrada (escala)
    llama a la función peticion_valor y obtiene el valor de temperatura que se quiere convertir
    que despues de validar si esta en rango se pasa mediante return
    Para validar llama la funcion conversor_to_c para comprobar si temperatura en celsius es nenor que -273.15 celsius
    """
 
    while True: 
        temp = peticion_valor()
        temperatura_celsius = coversor_to_c(temp, escala)

        if temperatura_celsius < -273.15:
            print("ATENCIÓN!! Valor de temperatura fuera de rango para la escala seleccionada: \n" \
                    "Min (C): -273.15, Min (K): 0, Min (F): -459.67, Min (Ra): 0, Min (Re): -218.52")               
        else:
            break              

    return temperatura_celsius
    

def esc_salida():
    ''' 
    Esta funcion no recibe parametros y sirve para personalizar las opciones de la funcion mostrar menu
    cada vez que se la llama  devuelve mediante return la escala seleccionada cada vez
    '''
    titulo_menu = "\n** Escalas de temperatura de salida a las que quieres convertir **"
    texto = "Elige las escalas de temperatura final que deseees (1-5) \n" \
    "       elige (6) para finalizar la selección: "
    opcion_fin = "6. Finalizar la selección de escalas de conversión \n"
    esc_final= mostrar_menu(titulo_menu, texto, opcion_fin)
    return esc_final


def bucle_esc_salida():
    ''' 
    Esta función no recibe parametros 
    llena la lista de tipo de escala de salida mediante un bucle, llama la funcion esc_salida
    que a su vez usa la funcion mostrar menu para retornar la lista de escalas a las que se desea convertir
    '''
    lista_esc_salida = []
    while True:
        escalas_salida = esc_salida()
        if escalas_salida == 6:
            break
        else:
            lista_esc_salida.append(escalas_salida)

    return lista_esc_salida


def coversor_to_c(temperatura, esc_ini):
    ''' 
    Esta función recibe por parametros el valor a convertir (temperatura) y la escala que es.
    Convierte cualquier valor a Celsius y lo retorna.
    '''

    if esc_ini == 1:
        celsius = temperatura                   # C Celsius to C Celsius

    elif esc_ini == 2:
        celsius = (temperatura-32)*5/9          # F Fahrenheit to C Celsius

    elif esc_ini == 3:
        celsius = temperatura - 273.15          # K Kelvin to C Celsius

    elif esc_ini == 4:
        celsius = (temperatura-491.67)*5/9      # Ra Rankine to C Celsius

    else:
        celsius = temperatura*5/4               # Re Réaumur to C Celsius

    return celsius


def conversor_from_c(temp_cel, esc_fin):
    ''' 
    Recibe por parametros la temperatura en Cenlsius y la lista con las escalas a las que se quiere convertir
    crea un diccionario que rellena con las conversiones de las escalas finales seleccionadas
    '''
    conversion = {}

    for escala in esc_fin:

        if escala == 1:
            temp_convertida = temp_cel                  # C Celsius to C Celsius
            conversion['Celsius'] = temp_convertida

        elif escala == 2:
            temp_convertida = (temp_cel*9/5)+32         # C celsius to F Fahrenheit
            conversion['Fahrenheit'] = temp_convertida

        elif escala == 3:
            temp_convertida = temp_cel + 273.15         # C Celsius to K Kelvin
            conversion['Kelvin'] = temp_convertida

        elif escala == 4:
            temp_convertida = (temp_cel+273.15)*9/5     # C Celsius to Ra Rankine
            conversion['Rankine'] = temp_convertida

        else:
            temp_convertida = temp_cel*4/5              # C Celsius to Re Réaumur
            conversion['Réaumur'] = temp_convertida

    return conversion
 
        
def mostrar_resultat(temperatura_final):
    ''' 
    Esta función recibe el diccionario conversion como argumentto (temperatura_final), 
    itera el diccionario y muestra meduiante print las temperaturas convertidas
    '''

    print(f"\nLa temperatura convertida es: ") 
    
    for escala, valor in temperatura_final.items():

        print(f"{escala}: {valor}")
    
    print("\nSi quieres salir de la Aplicación selecciona (9 + intro)")
    
def main():
    ''' 
    Es la función orquestadora que organiza el resto de funciones y
    el flujo de la App. 
    '''
    while True:                             # Mantiene la App activa hasta que usuario introduzca 9
        escala_ini = esc_entrada()          # Llamada a la función donde se selecciona la escala de entrada
        if escala_ini == 9:
            print("Saliendo de la App")
            break
        
        # Llamada a la función que valida el valor de temperatura a convertir
        # Desde dentro de esta función se llama a la función peticion_valor y la funcion conversor to C
        temperatura_cel = validacion_valor(escala_ini)      

        # Llamada a la funcion que genera un bucle para poder seleccionar varias escalas de salida
        escala_fin = bucle_esc_salida()

        # Llamada a la funcion que realiza la conversión de Celsius a las escalas de salida seleccionadas
        temp_fin = conversor_from_c(temperatura_cel, escala_fin)

        # Llamada a la función que muestra los resultados de la conversión
        mostrar_resultat(temp_fin)       

 
if __name__ == "__main__":
    main()
```

#### 3. Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

```python

def demanem_text():
    '''
    Sol·licita a l'usuari la introducció d'un text per consola.
    Controla que l'entrada no sigui buida ni contingui només espais en blanc.
    Manté un bucle fins que l'usuari introdueix un text vàlid.

    Returns:
        str: Text introduït per l'usuari, net d'espais als extrems.
    '''
    while True:
        text = input("Si us plau introdueix qualsevol text: ").strip()
        
        if text:
            return text
        else:
            print("Error: has d'introduir algun text, no pot estar buit.")
            

def netejar_frase(frase):
    '''
    Rep una frase i la neteja eliminant caràcters no alfabètics.
    Separa la frase en paraules i retorna una llista amb les paraules netes.

    Args:
        frase (str): Text d'entrada a processar.

    Returns:
        list: Llista de paraules només amb caràcters alfabètics.
    '''
    paraules = frase.split()
    paraules_netes = []

    for paraula in paraules:   
        paraula_neta = ""     
        for caracter in paraula:
            if caracter.isalpha():
                paraula_neta += caracter

        if paraula_neta:
            paraules_netes.append(paraula_neta)
    
    return paraules_netes


def paraules_count(paraules):
    '''
    Rep una llista de paraules i compta quantes vegades apareix cadascuna.

    Args:
        paraules (list): Llista de paraules netes.

    Returns:
        dict: Diccionari amb les paraules com a claus i el nombre d'aparicions com a valors.
    '''
    diccionari = {}

    for paraula in paraules:
        if paraula in diccionari:
            times = diccionari.get(paraula)
            times += 1
            diccionari[paraula] = times
        else: 
            diccionari[paraula] = 1

    return diccionari


def llargada_mitjana_paraules(paraules):
    '''
    Calcula la llargada mitjana de les paraules d'una llista.

    Args:
        paraules (list): Llista de paraules netes.

    Returns:
        float: Mitjana de caràcters per paraula (arrodonida a 2 decimals).
    '''
    suma_caracters = 0

    if len(paraules) > 0:
        for paraula in paraules:
            suma_caracters += len(paraula)

        mitjana = round(suma_caracters / len(paraules), 2)
    else:
        mitjana = 0

    return mitjana


def mostrar_resultat(diccionari_paraules):
    ''' 
    Mostra per pantalla el recompte de paraules.

    Rep un diccionari amb paraula → nombre d'aparicions i imprimeix
    cada element de forma clara.

    Args:
        diccionari_paraules (dict): Diccionari amb el recompte de paraules.
    '''
    print(f"\nEl recompte de paraules del text és: \n") 
    
    for paraula, valor in diccionari_paraules.items():
        print(f"{paraula}: {valor}")


def main():
    ''' 
    Funció principal orquestradora que coordina el flux de l'aplicació.

    Passos:
    1. Demana un text a l'usuari.
    2. Neteja la frase eliminant caràcters no alfabètics.
    3. Comptabilitza les paraules.
    4. Calcula la llargada mitjana de les paraules.
    5. Mostra els resultats per pantalla.
    '''
    frases = demanem_text()
        
    frase_neta = netejar_frase(frases.lower())

    comptar_paraules = paraules_count(frase_neta)

    mitjana_paraules = llargada_mitjana_paraules(frase_neta)

    print(f"\nAnàlisi del text proporcionat: {frases}")

    mostrar_resultat(comptar_paraules)

    print(f"\nLa llargada mitjana de les paraules del text és: {mitjana_paraules} lletres per paraula")


if __name__ == "__main__":
    main()
```

#### 4. Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

```python

def revertir_dic(dic_inicial):
    ''' 
    Recibe un diccionario y invierte las (clave,valor) convirtiendo los valores
    a claves y las claves a valores del diccionario retornado.
    Además si hay valores duplicados en el diccionario de entrada, los acumula en una lista 
    bajo la misma clave y imprime un mensaje de advertencia

    Args: 
        dic_inicial (diccionario): diccionario de entrada a procesar

    Returns:
        Diccionario claves = valores entrada si duplicados se presentan 
        con valores añadidos a lista | valores = lista de claves entrada
    '''
    dic_invertido = {}

    # bucle for para iterar el diccionario mediante método .items() que me devuelve tuplas "parejas" (clave,valor)

    for clave, valor in dic_inicial.items(): 

        # comprobar si valor ya existe en dic y sino crear la nueva clave

        if valor not in dic_invertido:
            dic_invertido[valor] = [clave]

        # si existe el valor, se agrega (clave) como valor en la lista para esa "clave" que viene del valor
        # quedando esa clave como una lista convarios valores
        # impresion de claves que tienen varios valores

        else:
            dic_invertido[valor].append(clave)
            print(f"Estos valores estan duplicados '-{valor}-'")

    return dic_invertido


dictionary_inicial = {
    'a': 'apple',
    'b': 'banana',
    'c': 'cherry',
    'd': 'apple',
    'e': 'banana',
    'f': 'date',
    'g': 'fig',
    'h': 'grape',
    'i': 'fig',
    'j': 'apple'
}

# Llamada a la funcion para revertir clave, valor del diccionario facilitado

diccionario_invertido = revertir_dic(dictionary_inicial)

#Imprimimos el diccionario_invertido resultante de la salida de la función

print(diccionario_invertido)
```
