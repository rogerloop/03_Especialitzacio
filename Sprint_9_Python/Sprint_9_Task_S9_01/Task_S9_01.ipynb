#### El cálculo del IMC

Para calcular el IMC es necesario dividir el peso en kg de un paciente entre el cuadrado de su altura en metros. Con esto obtendremos un valor que se situará en uno de los siguientes rangos:

Fórmula IMC = Peso Kg / (Altura m)**2

- Bajo peso: menos de 18,5 
- Peso normal: 18,5-24,9 
- Sobrepeso: 25-29,9 
- Obesidad: más de 30

```python
import numpy as np
import pandas as pd
```

```python
"""
Aplicación para calcular el IMC y categorizar el estado de salud
según el índice de masa corporal.
"""
# Funciones auxiliares
# _____________________________________________________
def peticion_datos(peticion):
    """
    Solicita al usuario los datos corporales a introducir y 
    retorna la variable respuesta que será validada fuera de la función
    """
    while True:
        try:
            entrada = input(peticion)
            # Sustituir coma por punto para evitar error al convertir STR a float
            entrada = entrada.replace(",", ".")
            respuesta = float(entrada)
            
            return respuesta

        except ValueError:            
            print("Entrada no válida. Introduce un número.")
    
def calculo_imc(peso, altura):
    """Esta función calcula y retorna el IMC
    índice de masa corporal, según los argumentos 
    peso y altura    
    """
    imc = round(peso / (altura**2),1)
    return imc

def categorizar(indice):
    """
    Esta función recibe un indice IMC y categoriza
    el estado en funcion de ese IMC, retornando la 
    categoria y una receta para cada estado
    """
    if indice < 18.5:
        cat = "bajo peso"
        receta = "sigue así tu salud lo agradecerá ;)"
    elif indice <= 24.9:
        cat = "peso normal"
        receta = "lo estás haciendo bién, mantén esta línea :)"
    elif indice <= 29.9:
        cat = "sobrepeso"
        receta = "cuidado, deberías hacer más ejercicio y vigilar la alimentación."
    else:
        cat = "obesidad"
        receta = "con este indice, estás poniendo en peligro tu vida, " \
                    "\ndeberías ir al médico con urgencia para encontrar solución"
    return(cat, receta)

def obtencion_datos_validados():
    """
    No recibe parametros, llama a la función peticion_datos dos veces
    y obtiene el pes y el alto que lo retorna mediante return
    """
    peticion_pes = "Por favor, ¿Nos puedes indicar tu peso en kilogramos (solo valor)? " \
                        "\nintroduce un peso coherente entre 30 y 250 Kg: "
    peticion_alto = "¿Nos puedes facilitar tu altura en metros (solo valor)? " \
                        "\nintroduce una altura coherente entre 1 y 2.5 m: "

    while True: 
        pes = peticion_datos(peticion_pes)      
        if 30 <= pes <= 250:
            break     
        else:
            print("Valor de peso incorrecto o fuera de rango")  

    while True:
        alto = peticion_datos(peticion_alto)
        if 1 <= alto <= 2.5:
            break
        else:
            print("Valor de altura incorrecto o fuera de rango")

    return(pes, alto)

# Función orquestadora MAIN para llamar al restio de funciones
#_______________________________________________________

def main():
        
    # mensaje para usuario explica lo que hará la app
    print("Vamos a analizar tu índice de masa corporal (IMC), " \
        "\npara poder calcularlo vamos a necesitar la siguiente información:")
    
    # llamamos funcion para obtener datos usuario validados y pasarlos al calculo
    pes, alto = obtencion_datos_validados()

    # pasamos los datos del usuario a la duncion calculo para obtener IMC
    calculo = calculo_imc(pes, alto)

    # pasamos a funcion categorizar el calculo IMC para que categorice
    categoria, propuesta = categorizar(calculo)

    # Respuesta al usuario del analisis y categorizacion de IMC
    print(f"Despúes de analizar tu peso de {pes} Kg. y altura de {alto} m, podemos afirmar que: \
        \nTu IMC es {calculo}, esto indica que estas en la categoría de {categoria},\
            \nnuestro consejo en funión de tu IMC es: {propuesta} ")

if __name__ == "__main__":
    main()
```

#### 2. Convertidor de temperatures

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

```python
from IPython.display import clear_output

def mostrar_menu(cabecera, texto, opcion):
    salir = False
  
    print(cabecera, "\n")
   
    while not salir:
        try:

            print(f'''Escala:
            1. Celsius (C)
                    
            2. Fahrenheit (F)
                    
            3. Kelvin (K)
                    
            4. Rankine (Ra)
                    
            5. Réamur (Re)   
                        
            {opcion}''')
        
            clear_output(wait=True)
    
            return int(input(texto))
         
        except Exception as e:
            print(f'Ocurrio un error: {e}')

# para toda la app para palabra escala usamos diminutivo (esc)


def esc_entrada():
    titulo_menu = "** Escalas de temperatura de entrada para convertir **"
    texto = "Elige la escala de temperatura inicial (1-5) u otra opción: "
    opcion_salir = "9. Salir para finalizar Aplicación"
    esc_inicial = mostrar_menu(titulo_menu, texto, opcion_salir)

    return esc_inicial


def peticion_valor():
    """
    Solicita al usuario los datos corporales a introducir y 
    retorna la variable respuesta que será validada fuera de la función
    """
    while True:
        try:
            entrada = input("Introduce el valor a convertir: ")
            # Sustituir coma por punto para evitar error al convertir STR a float
            entrada = entrada.replace(",", ".")
            valor = float(entrada)
            
            return valor

        except ValueError:            
            print("Entrada no válida. Introduce un número valido.")


def validacion_valor(escala):
    """
    No recibe parametros, llama a la función peticion_valor 
    y obtiene el valor de temperatura que se quiere convertir
    que despues de validar esta en rango se pasa mediante reurn
    """
 
    while True: 
        temp = peticion_valor()
        temperatura_celsius = coversor_to_c(temp, escala)

        if temperatura_celsius < -273.15:
            print("ATENCIÓN!! Valor de temperatura fuera de rango para la escala seleccionada: \n" \
            "Min (C): -273.15, Min (K): 0, Min (F): -459.67, Min (Ra): 0, Min (Re): -218.52")               
        else:
            break              

    return temperatura_celsius
    

def esc_salida():
    titulo_menu = "\n** Escalas de temperatura de salida a las que quieres convertir **"
    texto = "Elige la escala de temperatura final (1-5) u otra opción: "
    opcion_fin = "6. Finalizar la selección de escalas de conversión \n" \
                    "\n            9. Salir para finalizar Aplicación"
    esc_final= mostrar_menu(titulo_menu, texto, opcion_fin)
    return esc_final


def bucle_esc_salida():
    lista_esc_salida = []
    while True:
        escalas_salida = esc_salida()
        if escalas_salida == 6:
            break
        else:
            lista_esc_salida.append(escalas_salida)

    return lista_esc_salida


def coversor_to_c(temperatura, esc_ini):

    if esc_ini == 1:
        celsius = temperatura                   # C Celsius to C Celsius

    elif esc_ini == 2:
        celsius = (temperatura-32)*5/9          # F Fahrenheit to C Celsius

    elif esc_ini == 3:
        celsius = temperatura - 273.15          # K Kelvin to C Celsius

    elif esc_ini == 4:
        celsius = (temperatura-491.67)*5/9      # Ra Rankine to C Celsius

    else:
        celsius = temperatura*5/4               # Re Réaumur to C Celsius

    return celsius


def conversor_from_c(temp_cel, esc_fin):
    conversion = {}

    for escala in esc_fin:

        if escala == 1:
            temp_convertida = temp_cel                  # C Celsius to C Celsius
            conversion['Celsius'] = temp_convertida

        elif escala == 2:
            temp_convertida = (temp_cel*9/5)+32         # C celsius to F Fahrenheit
            conversion['Fahrenheit'] = temp_convertida

        elif escala == 3:
            temp_convertida = temp_cel + 273.15         # C Celsius to K Kelvin
            conversion['Kelvin'] = temp_convertida

        elif escala == 4:
            temp_convertida = (temp_cel+273.15)*9/5     # C Celsius to Ra Rankine
            conversion['Rankine'] = temp_convertida

        else:
            temp_convertida = temp_cel*4/5              # C Celsius to Re Réaumur
            conversion['Réaumur'] = temp_convertida

    return conversion
 
        
def mostrar_resultado(temperatura_final):

    print(f"La temperatura convertida es: {temperatura_final}") 
    
    
def main():
    while True:
        escala_ini = esc_entrada()
        if escala_ini == 9:
            print("Saliendo de la App")
            break
        
        temperatura_cel = validacion_valor(escala_ini)
        escala_fin = bucle_esc_salida()
        if escala_fin == 9:
            print("Saliendo de la App")
            break    
        temp_fin = conversor_from_c(temperatura_cel, escala_fin)

        mostrar_resultado(temp_fin)

 
if __name__ == "__main__":
    main()
```
