## Nivell 1

Des d’un Jupyter Notebook faràs els següents exercicis utilitzant la llibreria requests de Python.

```python
import requests
```

### 1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

- Llista de publicacions (/posts)
- Llista d’usuaris (/users)
- Llista de tasques (/todos)

```python
base_url = "https://jsonplaceholder.typicode.com"

# GET Posts

posts_response = requests.get(f"{base_url}/posts")

# GET Users

users_response = requests.get(f"{base_url}/users")

# GET Todos

todos_response = requests.get(f"{base_url}/todos")
```

### 2. Mostra per pantalla:

- La quantitat total de cada recurs.
- El codi d'estat de cada petició.

```python

# Convert JSON responses
posts = posts_response.json()
users = users_response.json()
todos = todos_response.json()

print("POSTS")
print(f"Total resources: {len(posts)}")
print(f"Status code: {posts_response.status_code}")

print("\nUSERS")
print(f"Total resources: {len(users)}")
print(f"Status code: {users_response.status_code}")

print("\nTODOS")
print(f"Total resources: {len(todos)}")
print(f"Status code: {todos_response.status_code}")
```

### 3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
not_found_response = requests.get(f"{base_url}/posts/999999")

print("NON-EXISTENT POST")
print(f"Status code: {not_found_response.status_code}")
```

### 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
new_post = {
    "title": "My First API Post",
    "body": "This is a test post created with Requests.",
    "userId": 1
}

post_response = requests.post(
    f"{base_url}/posts",
    json=new_post
)

print("POST REQUEST")
print(f"Status code: {post_response.status_code}")

print("\nJSON Response:")
print(post_response.json())
```

### 5. Fes una petició PATCH per modificar parcialment una publicació existent.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
updated_data = {
    "title": "Updated Title"
}

patch_response = requests.patch(
    f"{base_url}/posts/1",
    json=updated_data
)

print("PATCH REQUEST")
print(f"Status code: {patch_response.status_code}")

print("\nJSON Response:")
print(patch_response.json())
```

### 6. Fes una petició DELETE sobre una publicació.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
delete_response = requests.delete(
    f"{base_url}/posts/1"
)

print("DELETE REQUEST")
print(f"Status code: {delete_response.status_code}")

print("\nJSON Response:")
print(delete_response.json())
```

Open-Meteo es una API meteorológica gratuita que permite realizar peticiones GET para obtener información sobre previsiones meteorológicas, temperaturas, precipitaciones, velocidad del viento y otros datos climáticos.

La API devuelve los resultados en formato JSON y no requiere autenticación para consultas básicas.



### Ejercicio 2

#### Endpoint 1

Forecast API

https://api.open-meteo.com/v1/forecast

Permite obtener previsiones meteorológicas para una ubicación determinada.

---

#### Endpoint 2

Marine Weather API

https://marine-api.open-meteo.com/v1/marine

Permite consultar información meteorológica marina.

---

#### Parámetros opcionales interesantes

##### latitude

Latitud de la ubicación.

Ejemplo:

latitude=41.3888

---

##### longitude

Longitud de la ubicación.

Ejemplo:

longitude=2.159

---

##### current

Permite solicitar variables meteorológicas actuales.

Ejemplo:

current=temperature_2m

---

##### hourly

Permite solicitar variables horarias.

Ejemplo:

hourly=temperature_2m

---

#### Formato de respuesta

La API devuelve la información en formato JSON.

### 3. Fes una petició GET senzilla:

- Mostra el codi d'estat de la resposta.
- Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
import requests
import pandas as pd
```

```python
url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=41.3888"
    "&longitude=2.159"
    "&current=temperature_2m,relative_humidity_2m,wind_speed_10m"
)

response = requests.get(url)

print(f"Status code: {response.status_code}")
```

```python
data = response.json()

print("Latitude:", data["latitude"])
print("Longitude:", data["longitude"])

print("\nCurrent Weather Data")

print(
    f"Temperature: {data['current']['temperature_2m']} °C"
)

print(
    f"Relative Humidity: {data['current']['relative_humidity_2m']} %"
)

print(
    f"Wind Speed: {data['current']['wind_speed_10m']} km/h"
)
```

### 4. Converteix les dades a un DataFrame de pandas:

Mostra les primeres files del DataFrame.

```python
hourly_url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=41.3888"
    "&longitude=2.159"
    "&hourly=temperature_2m"
)

hourly_response = requests.get(hourly_url)

hourly_data = hourly_response.json()
```

```python
weather_df = pd.DataFrame({
    "time": hourly_data["hourly"]["time"],
    "temperature": hourly_data["hourly"]["temperature_2m"]
})

weather_df.head()
```

### 2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python
import requests
import pandas as pd
```

```python
search_url = (
    "https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_search"
)

params = {
    "q": "estacions meteorologiques"
}

search_response = requests.get(search_url, params=params)

print(f"Status code: {search_response.status_code}")
```

### 3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

Recuperar almenys 100 registres.

```python
search_data = search_response.json()

print("Datasets encontrados:")
print()

for dataset in search_data["result"]["results"][:5]:
    print(dataset["title"])
```

```python
dataset_id = search_data["result"]["results"][0]["name"]

show_url = (
    "https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_show"
)

show_response = requests.get(
    show_url,
    params={"id": dataset_id}
)

print(f"Status code: {show_response.status_code}")
```

```python
dataset_info = show_response.json()

print("Título:")
print(dataset_info["result"]["title"])

print("\nNúmero de recursos:")
print(len(dataset_info["result"]["resources"]))
```

### 4. Converteix els resultats a un DataFrame.

```python
csv_url = (
    "https://opendata-ajuntament.barcelona.cat/data/"
    "dataset/feafec8a-b389-42b5-a85d-cf16f3976440/resource/"
    "feafec8a-b389-42b5-a85d-cf16f3976440/download/"
    "2020_meteocat_estacions.csv"
)

df = pd.read_csv(csv_url)
```
