```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("penguins")
# df = sns.load_dataset("iris")
# df = sns.load_dataset("titanic")


def estilo_frafico(col):

    plt.figure(figsize=(5,5))

    plt.title(f"Type: {df[col].dtype} - Columna: {col}")

    plt.xlabel(col)
    plt.ylabel("Frecuencia")

    plt.tight_layout()

def histograma(df,col):

    estilo_frafico(col)
    plt.title(f"Histograma de la columna: {col}\nType: {df[col].dtype}")
    df[col].plot(kind="hist", bins=15)    
    plt.show()


def grboxplot(df, col):

    estilo_frafico(col)
    plt.title(f"Boxplot de la columna: {col}\nType: {df[col].dtype}")
    df[col].plot(kind="box")
    plt.show()


def barres(df, col):
        
    estilo_frafico(col)
    plt.title(f"Gráfico de barras, columna: {col}\nType: {df[col].dtype}")
    df[col].value_counts().head(10).plot(kind="bar")
    plt.show()


def pastel(df, col):
    
    estilo_frafico(col)
    plt.title(f"Pstel de la columna: {col}\nType: {df[col].dtype}")
    df[col].value_counts().plot(kind="pie")    
    plt.show()
    

def barres_any(df, col):      
    
    estilo_frafico(col)
    plt.title(f"Años más frecuentes: {col}\nType: {df[col].dtype}")
    df[col].dt.year.value_counts().sort_index().plot(kind="bar")  
    plt.show()


def elegir_grafico (df):

    for col in df.columns:

        if (df[col].dtype == "float" or  df[col].dtype == "int"):

            if df[col].nunique() < 5:

                grboxplot(df, col)  

            else:

                histograma(df, col) 

        elif df[col].dtype == "str":
            
            barres(df, col)

        elif df[col].dtype == "category":
          
           barres(df, col)
        
        elif df[col].dtype == "bool":

            pastel(df,col)

        elif df[col].dtype == "datetime" or df[col].dtype == "datetime64" or df[col].dtype == "datetime64[us]":

            barres_any(df,col)
            
        else:
            
            print('otros tipos ', df[col].dtype)

            

elegir_grafico(df)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# ==========================================
# DATAFRAME DE EJEMPLO
# ==========================================

df = sns.load_dataset("titanic")

# También puedes probar:
# df = sns.load_dataset("penguins")
# df = sns.load_dataset("iris")

df = treballadors_2025
# ==========================================
# FUNCIÓN PRINCIPAL
# ==========================================

def generar_graficos(df):

    # Recorremos todas las columnas
    for columna in df.columns:

        # Creamos figura nueva
        plt.figure(figsize=(8, 5))

        # Guardamos el dtype como string
        tipo_dato = str(df[columna].dtype)

        # Número de valores únicos
        cantidad_unicos = df[columna].nunique()

        # ==========================================
        # VARIABLES NUMÉRICAS
        # ==========================================

        if "int" in tipo_dato or "float" in tipo_dato:

            # --------------------------------------
            # NUMÉRICA BINARIA
            # Ejemplo: 0/1
            # --------------------------------------

            if cantidad_unicos == 2:

                df[columna].value_counts().sort_index().plot(kind="bar")

                plt.title(
                    f"Variable numérica binaria: {columna}\n(dtype: {tipo_dato})"
                )

                plt.xlabel(columna)
                plt.ylabel("Frecuencia")


            # --------------------------------------
            # NUMÉRICA CON POCOS VALORES
            # → Boxplot
            # --------------------------------------

            elif cantidad_unicos < 10:

                df[columna].plot(kind="box")

                plt.title(
                    f"Boxplot de la columna: {columna}\n(dtype: {tipo_dato})"
                )

                plt.ylabel(columna)


            # --------------------------------------
            # NUMÉRICA ESTÁNDAR
            # → Histograma
            # --------------------------------------

            else:

                df[columna].plot(kind="hist", bins=15)

                plt.title(
                    f"Histograma de la columna: {columna}\n(dtype: {tipo_dato})"
                )

                plt.xlabel(columna)
                plt.ylabel("Frecuencia")


        # ==========================================
        # VARIABLES CATEGÓRICAS / TEXTO
        # ==========================================

        elif (
            tipo_dato == "object"
            or tipo_dato == "category"
            or tipo_dato == "str"
        ):

            # --------------------------------------
            # CATEGÓRICA BINARIA
            # Ejemplo:
            # male/female
            # yes/no
            # --------------------------------------

            if cantidad_unicos == 2:

                df[columna].value_counts().plot(
                    kind="pie",
                    autopct="%1.1f%%"
                )

                plt.title(
                    f"Variable categórica binaria: {columna}\n(dtype: {tipo_dato})"
                )

                plt.ylabel("")


            # --------------------------------------
            # CATEGÓRICA NORMAL
            # --------------------------------------

            else:

                df[columna].value_counts().head(10).plot(kind="bar")

                plt.title(
                    f"Gráfico de barras: {columna}\n(dtype: {tipo_dato})"
                )

                plt.xlabel(columna)
                plt.ylabel("Frecuencia")


        # ==========================================
        # VARIABLES BOOLEANAS
        # ==========================================

        elif tipo_dato == "bool":

            df[columna].value_counts().plot(
                kind="pie",
                autopct="%1.1f%%"
            )

            plt.title(
                f"Variable booleana: {columna}\n(dtype: {tipo_dato})"
            )

            plt.ylabel("")


        # ==========================================
        # VARIABLES FECHA
        # ==========================================

        elif "datetime" in tipo_dato:

            # Extraemos años
            años = df[columna].dt.year

            años.value_counts().sort_index().plot(kind="bar")

            plt.title(
                f"Años más frecuentes: {columna}\n(dtype: {tipo_dato})"
            )

            plt.xlabel("Año")
            plt.ylabel("Frecuencia")


        # ==========================================
        # OTROS TIPOS
        # ==========================================

        else:

            print(f"No se pudo generar gráfico para: {columna}")
            print(f"Tipo detectado: {tipo_dato}")

            continue


        

        # Mostrar gráfico
        plt.show()


# ==========================================
# EJECUCIÓN
# ==========================================

generar_graficos(df)
```

#### Exercici 2

Carrega l'arxiu matriu_distancies.xlsx a pandas, de manera que els noms de files i els noms de columnes siguin els de les ciutats. Borra "Las Palmas de Gran Canaria" i "Palma" perquè poguem fer el trajecte en cotxe.

Ens interessa visitar totes les ciutats principals d'Espanya recorrent la mínima distància possible.

No cal que ho facis de forma òptima, ens interessa que desenvolupis una solució raonable utilitzant les eines que tens actualment.

Per exemple, una aproximació senzilla (que no òptima) seria anant sempre a la ciutat més propera que no haguem visitat encara

Fes una funció que donada la matriu de distàncies i la ciutat d'origen, faci una proposta de ruta que sigui el més curta possible que puguis, retornant una llista amb l'ordre de visita. Dóna també la distància total recorreguda.

(EXTRA) Des de quina ciutat la ruta seria més curta amb l'algoritme plantejat

```python
# IMportar las librerias de google OR-Tools
from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp

# ruta del fichoero a cargar (solo ruta sin nombre fichero)
ruta_file = "/Users/rogerdefez/Documents/Cursos i Llibres/BootCamp IT Academy/03_Especialitzacio/Sprint_10_Analitica_dades_Python/Datos originales/"

# carga fichero al dataframe
df = pd.read_excel(ruta_file + "matriu_distancies.xlsx", index_col=[0])

# Borrado ciudades Palma y Las Palmas de Gran Canaria para ir en coche
df = df.drop(["Palma","Las Palmas de Gran Canaria"]) # Borrado filas
df = df.drop(["Palma","Las Palmas de Gran Canaria"], axis=1)
df

# Convertir dataframe en la lista de valores (Matriz de distancias que necesita OR-Tools)
matriz_distancias = df.values.tolist()

# Convertir indice dataframe en lista ciudades
ciudades = df.index.tolist()


#------------------------------------------------------------------------------
# Defino funcion para crear el modelo de datos para OR-Tools
#------------------------------------------------------------------------------
def create_data_model():
    "Guarda los datos para el analisis"
    data = {}
    data["distance_matrix"] = matriz_distancias
    data["num_vehicles"] = 1
    data["depot"] = 0
    return data


#------------------------------------------------------------------------------
# Crea la devolución de llamada de distancia
#------------------------------------------------------------------------------
def distance_callback(from_index, to_index):
    """Returns the distance between the two nodes.
    toma cualquier par de ubicaciones y muestre la distancia entre ellas"""
    # Convert from routing variable Index to distance matrix NodeIndex.
    from_node = manager.IndexToNode(from_index)
    to_node = manager.IndexToNode(to_index)
    return df["distance_matrix"][from_node][to_node]

#------------------------------------------------------------------------------
# crea el administrador de índice (manager) y el modelo de enrutamiento (routing)
#------------------------------------------------------------------------------

df = create_data_model()
manager = pywrapcp.RoutingIndexManager(
    len(df["distance_matrix"]), df["num_vehicles"], df["depot"]
)
routing = pywrapcp.RoutingModel(manager)

#------------------------------------------------------------------------------
# Llama al evaluador de la curva de costes "transit_callback_index" En este caso el coste de viaje es
# solamente la distancia entre las 2 ciudades que devuelve "distance_callback".
#------------------------------------------------------------------------------
transit_callback_index = routing.RegisterTransitCallback(distance_callback)

routing.SetArcCostEvaluatorOfAllVehicles(transit_callback_index)

#------------------------------------------------------------------------------
# Establecer parámetros de búsqueda 
# El siguiente código establece los parámetros de búsqueda predeterminados 
# y un método heurístico para encontrar la primera solución:
#------------------------------------------------------------------------------

search_parameters = pywrapcp.DefaultRoutingSearchParameters()
search_parameters.first_solution_strategy = (
    routing_enums_pb2.FirstSolutionStrategy.PATH_CHEAPEST_ARC
)


#Agregar la impresora de la solución

# A continuación, se muestra la función que muestra la solución que muestra el solucionador. 
# La función extrae la ruta de la solución y la imprime en la consola.
# La función muestra la ruta óptima y su distancia, que proporciona ObjectiveValue().

def print_solution(manager, routing, solution):
    """Prints solution on console."""
    print(f"Objetivo: {solution.ObjectiveValue()} km")
    index = routing.Start(0)
    plan_output = "Ruta para el covhe 0:\n"
    route_distance = 0
    while not routing.IsEnd(index):
        plan_output += f" {manager.IndexToNode(index)} ->"
        previous_index = index
        index = solution.Value(routing.NextVar(index))
        route_distance += routing.GetArcCostForVehicle(previous_index, index, 0)
    plan_output += f" {manager.IndexToNode(index)}\n"
    print(plan_output)
    plan_output += f"Distancia de la ruta: {route_distance}Km\n"



# Resolver e imprimir la solución

# Llama al agente de resolución y, luego, imprimir la solución:

solution = routing.SolveWithParameters(search_parameters)
if solution:
    print_solution(manager, routing, solution)


def get_routes(solution, routing, manager):
    """Get vehicle routes from a solution and store them in an array."""
    # Get vehicle routes and store them in a two dimensional array whose
    # i,j entry is the jth location visited by vehicle i along its route.
    routes = []
    for route_nbr in range(routing.vehicles()):
        index = routing.Start(route_nbr)
        route = [manager.IndexToNode(index)]
        while not routing.IsEnd(index):
            index = solution.Value(routing.NextVar(index))
            route.append(manager.IndexToNode(index))
        routes.append(route)
    return routes



# 
routes = get_routes(solution, routing, manager)
# Display the routes.
for i, route in enumerate(routes):
  print('Route', i, route)
```

```python
# ==========================================================
# IMPORTACIÓN DE LIBRERÍAS
# ==========================================================

import pandas as pd

from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp


# ==========================================================
# CARGA Y LIMPIEZA DEL DATAFRAME
# ==========================================================

def cargar_y_limpiar_datos(ruta_excel):
    """
    Carga el archivo Excel de distancias entre ciudades,
    elimina ciudades no accesibles en coche y limpia
    los datos para OR-Tools.

    Args:
        ruta_excel (str):
            Ruta completa del archivo Excel.

    Returns:
        tuple:
            - matriz_distancias (list)
            - ciudades (list)
    """

    # ---------------------------------------------
    # Cargar dataframe
    # ---------------------------------------------

    df = pd.read_excel(
        ruta_excel,
        index_col=0
    )

    # ---------------------------------------------
    # Eliminar ciudades no accesibles en coche
    # ---------------------------------------------

    ciudades_eliminar = [
        "Palma",
        "Las Palmas de Gran Canaria"
    ]

    df = df.drop(
        index=ciudades_eliminar,
        columns=ciudades_eliminar
    )

    # ---------------------------------------------
    # Limpiar valores
    # OR-Tools trabaja mejor con enteros puros
    # ---------------------------------------------

    df = df.fillna(0)

    df = df.astype(int)

    # ---------------------------------------------
    # Convertir dataframe a estructuras Python
    # ---------------------------------------------

    matriz_distancias = df.values.tolist()

    ciudades = df.index.tolist()

    return matriz_distancias, ciudades


# ==========================================================
# CREAR DATA MODEL
# ==========================================================

def crear_data_model(
    matriz_distancias,
    inicio,
    final
):
    """
    Crea el diccionario de datos para OR-Tools.

    Args:
        matriz_distancias (list):
            Matriz cuadrada de distancias.

        inicio (int):
            Índice ciudad inicio.

        final (int):
            Índice ciudad final.

    Returns:
        dict:
            Configuración del problema.
    """

    data = {

        "distance_matrix": matriz_distancias,

        "num_vehicles": 1,

        "depot": inicio,

        "depot_final": final
    }

    return data


# ==========================================================
# CALLBACK DE DISTANCIA
# ==========================================================

def crear_callback_distancia(
    data,
    manager
):
    """
    Crea la función callback que OR-Tools utiliza
    para calcular distancias entre nodos.

    Args:
        data (dict):
            Diccionario del modelo.

        manager:
            RoutingIndexManager de OR-Tools.

    Returns:
        function:
            Función callback.
    """

    def distance_callback(
        from_index,
        to_index
    ):

        from_node = manager.IndexToNode(
            from_index
        )

        to_node = manager.IndexToNode(
            to_index
        )

        return int(
            data["distance_matrix"]
            [from_node]
            [to_node]
        )

    return distance_callback


# ==========================================================
# EXTRAER RUTA Y DISTANCIA
# ==========================================================

def obtener_ruta_y_distancia(
    solution,
    routing,
    manager,
    ciudades
):
    """
    Extrae la ruta calculada y la distancia total.

    Args:
        solution:
            Solución OR-Tools.

        routing:
            RoutingModel.

        manager:
            RoutingIndexManager.

        ciudades (list):
            Lista nombres ciudades.

    Returns:
        tuple:
            - ruta (list)
            - distancia_total (int)
    """

    index = routing.Start(0)

    ruta = []

    distancia_total = 0

    while not routing.IsEnd(index):

        nodo = manager.IndexToNode(index)

        ruta.append(ciudades[nodo])

        previous_index = index

        index = solution.Value(
            routing.NextVar(index)
        )

        distancia_total += (
            routing.GetArcCostForVehicle(
                previous_index,
                index,
                0
            )
        )

    nodo_final = manager.IndexToNode(index)

    ruta.append(ciudades[nodo_final])

    return ruta, distancia_total


# ==========================================================
# RESOLVER TSP (Travelling Salesman Problem)
# ==========================================================

def resolver_tsp(
    matriz_distancias,
    ciudades,
    ciudad_inicio,
    ciudad_final
):
    """
    Resuelve el problema TSP para una combinación
    concreta de ciudad inicio y final.

    Args:
        matriz_distancias (list):
            Matriz de distancias.

        ciudades (list):
            Lista nombres ciudades.

        ciudad_inicio (str):
            Ciudad origen.

        ciudad_final (str):
            Ciudad destino final.

    Returns:
        tuple:
            - ruta (list)
            - distancia (int)
    """

    # ---------------------------------------------
    # Convertir nombres a índices
    # ---------------------------------------------

    inicio = ciudades.index(
        ciudad_inicio
    )

    final = ciudades.index(
        ciudad_final
    )

    # ---------------------------------------------
    # Crear modelo datos
    # ---------------------------------------------

    data = crear_data_model(
        matriz_distancias,
        inicio,
        final
    )

    # ---------------------------------------------
    # Crear manager
    # ---------------------------------------------

    manager = pywrapcp.RoutingIndexManager(
        len(data["distance_matrix"]),
        data["num_vehicles"],
        [data["depot"]],
        [data["depot_final"]]
    )

    # ---------------------------------------------
    # Crear routing model
    # ---------------------------------------------

    routing = pywrapcp.RoutingModel(
        manager
    )

    # ---------------------------------------------
    # Registrar callback distancias
    # ---------------------------------------------

    distance_callback = (
        crear_callback_distancia(
            data,
            manager
        )
    )

    transit_callback_index = (
        routing.RegisterTransitCallback(
            distance_callback
        )
    )

    routing.SetArcCostEvaluatorOfAllVehicles(
        transit_callback_index
    )

    # ---------------------------------------------
    # Parámetros búsqueda
    # ---------------------------------------------

    search_parameters = (
        pywrapcp.DefaultRoutingSearchParameters()
    )

    search_parameters.first_solution_strategy = (
        routing_enums_pb2
        .FirstSolutionStrategy
        .PATH_CHEAPEST_ARC
    )

    # ---------------------------------------------
    # Resolver problema
    # ---------------------------------------------

    solution = routing.SolveWithParameters(
        search_parameters
    )

    # ---------------------------------------------
    # Obtener resultados
    # ---------------------------------------------

    if solution:

        return obtener_ruta_y_distancia(
            solution,
            routing,
            manager,
            ciudades
        )

    return None, None


# ==========================================================
# BUSCAR MEJOR RUTA GLOBAL
# ==========================================================

def buscar_mejor_ruta(
    matriz_distancias,
    ciudades
):
    """
    Prueba todas las combinaciones posibles
    de inicio y final.

    Guarda la mejor ruta encontrada.

    Args:
        matriz_distancias (list):
            Matriz distancias.

        ciudades (list):
            Lista nombres ciudades.

    Returns:
        dict:
            Información mejor ruta.
    """

    mejor_resultado = {

        "ruta": None,

        "distancia": float("inf"),

        "inicio": None,

        "final": None
    }

    # ---------------------------------------------
    # Probar todas las combinaciones
    # ---------------------------------------------

    for ciudad_inicio in ciudades:

        for ciudad_final in ciudades:

            if ciudad_inicio != ciudad_final:

                print("=" * 80)

                print(
                    f"Probando ruta desde "
                    f"{ciudad_inicio} "
                    f"hasta "
                    f"{ciudad_final}"
                )

                ruta, distancia = resolver_tsp(
                    matriz_distancias,
                    ciudades,
                    ciudad_inicio,
                    ciudad_final
                )

                if ruta:

                    print(
                        f"Distancia total: "
                        f"{distancia} km"
                    )

                    print(
                        " -> ".join(ruta)
                    )

                    # ---------------------------------
                    # Guardar mejor resultado
                    # ---------------------------------

                    if distancia < mejor_resultado["distancia"]:

                        mejor_resultado["ruta"] = ruta

                        mejor_resultado["distancia"] = distancia

                        mejor_resultado["inicio"] = ciudad_inicio

                        mejor_resultado["final"] = ciudad_final

    return mejor_resultado


# ==========================================================
# FUNCIÓN PRINCIPAL ORQUESTADORA
# ==========================================================

def main():

    """
    Función principal de ejecución.
    """

    ruta_excel = (
        "/Users/rogerdefez/Documents/"
        "Cursos i Llibres/"
        "BootCamp IT Academy/"
        "03_Especialitzacio/"
        "Sprint_10_Analitica_dades_Python/"
        "Datos originales/"
        "matriu_distancies.xlsx"
    )

    # ---------------------------------------------
    # Cargar datos
    # ---------------------------------------------

    matriz_distancias, ciudades = (
        cargar_y_limpiar_datos(
            ruta_excel
        )
    )

    # ---------------------------------------------
    # Buscar mejor ruta
    # ---------------------------------------------

    mejor_resultado = buscar_mejor_ruta(
        matriz_distancias,
        ciudades
    )

    # ---------------------------------------------
    # Mostrar mejor resultado
    # ---------------------------------------------

    print("\n" + "#" * 80)

    print("MEJOR RUTA ENCONTRADA")

    print("#" * 80)

    print(
        f"\nCiudad inicio: "
        f"{mejor_resultado['inicio']}"
    )

    print(
        f"Ciudad final: "
        f"{mejor_resultado['final']}"
    )

    print(
        f"\nDistancia mínima: "
        f"{mejor_resultado['distancia']} km"
    )

    print("\nRuta final:")

    print(
        " -> ".join(
            mejor_resultado["ruta"]
        )
    )


# ==========================================================
# EJECUCIÓN APP
# ==========================================================

if __name__ == "__main__":

    main()
```
