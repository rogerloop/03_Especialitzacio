#### Exercici 2

- Crea una columna que sigui el nom complet.
- Crea una columna si la persona és nascuda a Espanya o no.
- Posa el DNI com a índex del DataFrame (noms de files).
- Substitueix el nom de les columnes Dia de Naixement, Mes de Naixement i Any de Naixement per Dia, Mes i Any.
- Substitueix H per Home, D per Dona, A per Altres i NC per una dada faltant (nan/null/na).

```python
# 1. - Crea una columna que sigui el nom complet.
# Nombre completo con concat

treballadors_2025['Nom Complet'] = treballadors_2025['Nom'] + ' ' + treballadors_2025['Cognoms']

treballadors_2025
```

```python
# 2. - Crea una columna si la persona és nascuda a Espanya o no.

treballadors_2025['Nascut_Espanya'] = np.where(treballadors_2025["País d'origen"] == 'Espanya', 'Si', 'No')
treballadors_2025.head(100)
```

```python
treballadors_2025.describe()
```

```python
# 3. - Posa el DNI com a índex del DataFrame (noms de files).
# Con set_index() indico la columna que quiero asignar como indice del df, tambien podría usar inplace=True para machacar el propio df
# treballadors_2025.set_index('DNI', inplace=True)

treballadors_2025 = treballadors_2025.set_index('DNI')
treballadors_2025
```

```python
# 4. - Substitueix el nom de les columnes Dia de Naixement, Mes de Naixement i Any de Naixement per Dia, Mes i Any.
# Uso metodo rename junto al diccionario de nombre columnas antes:despues

treballadors_2025 = treballadors_2025.rename(columns={"Dia de Naixement": "Dia", "Mes de Naixement": "Mes", "Any de Naixement": "Any"})
print(treballadors_2025.columns)
```

```python
# 5. - Substitueix H per Home, D per Dona, A per Altres i NC per una dada faltant (nan/null/na).
# el metode replace() permet canviar un valor per un altre, en aquest cas com que eren varis he fet servir un diccionari
#  uso np.nan ya que es el objeto nulo de python es un valor especial de ausencia de datos

treballadors_2025['Gènere'] = treballadors_2025['Gènere'].replace({
    'H': 'Home',
    'D': 'Dona',
    'A': 'Altres',
    'NC': np.nan
})

treballadors_2025.tail(100)
```

```python
# 6- Mostra tots els canvis que has realitzat en una sola taula.

cambios = treballadors_2025[['Nom Complet', 'Nascut_Espanya', 'Dia', 'Mes', 'Any', 'Gènere']]
cambios
```

#### Exercici 3

Junta les columnes Fills i No Fills en una sola columna, utilitzant el mètode .apply() i definint una funció que resolgui el problema. La columna nova ha de dir-se "Fills" i prendre els valors "Sí" o "No".

- Para solucionar problema planteo que si tiene Fills = 1 obliga a "No Fills" = 0  y viceversa, 
así que en la funcion solo tengo que averiguar si Fills = 1 
- con la respuesta de la funcion 'tiene_hijos' lleno la columna Fills y despues elimino con Drop la 'No Fills'




```python
def tiene_hijos(fill):
    if fill == 1:
        return "Sí"
    else:
        return "No"
    
treballadors_2025['Fills'] = treballadors_2025['Fills'].apply(tiene_hijos)
treballadors_2025 = treballadors_2025.drop(['No Fills'], axis=1)

treballadors_2025
```

#### Exercici 4

- Crea una taula resum que permeti veure el sou mig, medià, mínim i màxim per Gènere.
- Ordena la taula en funció del sou mig.


- Afrontaré este reto con un Group by Gènere y luego con una agregacion multiple finalmente el Order By Sueldo Medio
- Reto encontrado es que 'Salari mensual' viene como string y hay que transformar a int

```python
treballadors_2025['Salari mensual'] = (treballadors_2025['Salari mensual']
                                    .str.replace(".","", regex=False)
                                    .str.replace("€","", regex=False)
                                    .str.replace(" ","", regex=False)
                                    .astype(int))

resum = (treballadors_2025.groupby(['Gènere'])['Salari mensual']
         .agg(['mean', 'median', 'min', 'max'])
         .sort_values('mean', ascending=False)
         .round(2)
         )
resum
```

#### Exercici 5

- Crea una taula resum amb el salari mig per gènere (files) i país d'origen (columnes).
- Afegeix-hi les mitjanes als marges de la taula.
- (EXTRA): Aplica format condicional a la taula per veure en un color més intens els valors més elevats

```python
treballadors_2025.info()
```

```python
resum = pd.crosstab(treballadors_2025["Gènere"], 
            treballadors_2025["País d'origen"], 
            values= treballadors_2025["Salari mensual"], 
            aggfunc='mean',
            margins=True, margins_name="Total").round(2)

resum.style.apply(lambda x: ['background-color: yellow' if v == x.max() else '' for v in x]).format('{:.2f}')
```
