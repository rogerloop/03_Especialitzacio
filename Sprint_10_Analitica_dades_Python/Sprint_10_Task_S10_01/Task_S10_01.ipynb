```python
#  Importar librerias

import pandas as pd

# 1. Ajustar ruta del excel
ruta_datos = '/Users/rogerdefez/Documents/Cursos i Llibres/BootCamp IT Academy/03_Especialitzacio/Sprint_10_Analitica_dades_Python/Datos originales/sprint10.xlsx'

# Cargar el dataframe...(df) (Importar), la hoja Sheet1 y convierto indice del excel en indice del Dataframe (evito duplicar)
treballadors_2025 = pd.read_excel(ruta_datos, sheet_name="Sheet1", header=3, index_col=0) 

# 2. Ordeno el df por país y ciudad con sort_values que primero ordena por columna1 y despues por columna2
#  reseteo el indice sobre df y con drop evito me cree columna 
treballadors_2025 = treballadors_2025.sort_values(by=["País d'origen", "Ciutat"]).reset_index(drop=True)

# 3. Muestro con head las 10 primereas filas
treballadors_2025.head(10)

# 4. Con is_unique compruebo dinamicamente que el df no tiene DNI duplicados (True) 
# o si tiene DNI duplicados, print con resultado

if treballadors_2025['DNI'].is_unique:
    print("No existen DNI duplicados en el dataframe treballadors_2025")

else: 
    print("Existen DNI duplicados en el dataframe treballadors_2025")
```
