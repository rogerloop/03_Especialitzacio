#### Exercici 6
- Crea una columna nova que sigui la data de naixament en format Datetime a partir de les columnes dia, mes i any. Utilitzant aquesta columna crea una funció que donada una data, et calculi l'edat actual a dia d'avui.
- Utilitza la funció que acabes de crear per generar una columna nova al DataFrame amb l'edat actual.


```python
# 1- Creo columna y convierto a datetime
treballadors_2025['Data_naixament'] = pd.to_datetime({'year': treballadors_2025['Any'],
                                                    'month': treballadors_2025['Mes'],
                                                    'day': treballadors_2025['Dia']})

treballadors_2025
```

```python
# 2- Creacion funcion calculadora de edad
def calcular_edad(naixament):
    
    hoy = datetime.today()
    
    # uso .days en lugar de dt.days porque .apply me manda un valor cada vez, no una serie
    return (hoy - naixament).days // 365

treballadors_2025['Edat_actual'] = treballadors_2025['Data_naixament'].apply(calcular_edad)
treballadors_2025
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset("titanic")
df

def estilo_frafico(col):

    plt.title(f"Type: {df[col].dtype} - Columna: {col}")

    plt.xlabel(col)
    plt.ylabel("Frecuencia")

    plt.tight_layout()

def histograma(df,col):

    plt.figure()
 
    df[col].plot(kind="hist")
    estilo_frafico(col)
    plt.show()


def grboxplot(df, col):

    plt.figure()
    plt.title(f"Type: {df[col].dtype} - Columna: {col}")
    df[col].plot(kind="box")
    plt.show()


def barres(df, col):
    
    plt.figure()
    plt.title(f"Type: {df[col].dtype} - Columna: {col}")
    df[col].value_counts().plot(kind="bar")
    plt.show()


def pastel(df, col):

    plt.figure()
    
    df[col].value_counts().plot(kind="pie")
    estilo_frafico(col)
    plt.show()
    

def barres_any(df, col):      

    plt.figure()
    plt.title(f"Type: {df[col].dtype} - Columna: {col}")
    df[col].value_counts().plot(kind="bar")  
    plt.show()


def elegir_grafico (df):

    for col in df.columns:

        if (df[col].dtype == "float" or  df[col].dtype == "int"):

            #if (df[col].dtype == "float" or  df[col].dtype == "int"):

            if df[col].nunique() < 5:

                grboxplot(df, col)  

            else:

                histograma(df, col) 

        elif df[col].dtype == "str":
            
            barres(df, col)

        elif df[col].dtype == "category":
          
           barres(df, col)
        
        elif df[col].dtype == "bool":

            pastel(df,col)

        elif df[col].dtype == "datetime" or df[col].dtype == "datetime64" or df[col].dtype == "datetime64[us]":

            barres_any(df,col)
            
        else:
            
            print('otros tipos ', df[col].dtype)

            

elegir_grafico(df)
```
