### Opción A

Solución 100% mia para mostrar gráficos

Sobre el Dataframe que hemos elegido en Celda anterior para dibujar sus columnas:

- Organizo la solución con funciones para dibujar cada tipo de gráfico
- Creo una función para mantener el estilo de grafico lo más unificado posible además de evitar repetir códigop
- Creo una función elegir gráfico que decide que grafico dibujar en funcion de:

- - el tipo de datos que contiene 
- - la cantidad de datos únicos que tiene (no su repetición)

```python

#----------------------------
# Creación estilo comun
#----------------------------

def estilo_frafico(col):

    """
    Aplica un estilo gráfico común para todos los gráficos.

    Configura:
    - tamaño de figura
    - título genérico
    - etiquetas de ejes
    - ajuste automático de márgenes

    Args:
        col (str):
            Nombre de la columna del dataframe.
    """

    plt.figure(figsize=(5,5))

    plt.title(f"Type: {df[col].dtype} - Columna: {col}")

    plt.xlabel(col)
    plt.ylabel("Frecuencia")

    plt.tight_layout()


#----------------------------
# Tipos de grafico a mostrar
#----------------------------

def histograma(df,col):

    """
    Genera un histograma para columnas numéricas
    con muchos valores únicos.

    Args:
        df (DataFrame):
            DataFrame que contiene los datos.

        col (str):
            Nombre de la columna a graficar.
    """

    estilo_frafico(col)
    plt.title(f"Histograma de la columna: {col}\nType: {df[col].dtype}")
    df[col].plot(kind="hist", bins=15)    
    plt.show()


def grboxplot(df, col):

    """
    Genera un boxplot para columnas numéricas
    con pocos valores únicos.

    Args:
        df (DataFrame):
            DataFrame que contiene los datos.

        col (str):
            Nombre de la columna a graficar.
    """

    estilo_frafico(col)
    plt.title(f"Boxplot de la columna: {col}\nType: {df[col].dtype}")
    df[col].plot(kind="box")
    plt.show()


def barres(df, col):

    """
    Genera un gráfico de barras para variables
    categóricas mostrando los valores más frecuentes.

    Args:
        df (DataFrame):
            DataFrame que contiene los datos.

        col (str):
            Nombre de la columna a graficar.
    """

    estilo_frafico(col)
    plt.title(f"Gráfico de barras, columna: {col}\nType: {df[col].dtype}")
    df[col].value_counts().head(10).plot(kind="bar")
    plt.show()


def pastel(df, col):
     
    """
    Genera un gráfico de pastel para variables
    binarias o booleanas.

    Args:
        df (DataFrame):
            DataFrame que contiene los datos.

        col (str):
            Nombre de la columna a graficar.
    """   

    estilo_frafico(col)
    plt.title(f"Pstel de la columna: {col}\nType: {df[col].dtype}")
    df[col].value_counts().plot(kind="pie", autopct="%1.1f%%")    
    plt.show()
    

def barres_any(df, col):

    """
    Genera un gráfico de barras agrupando
    frecuencias por año para columnas datetime.

    Args:
        df (DataFrame):
            DataFrame que contiene los datos.

        col (str):
            Nombre de la columna datetime.
    """
        
    estilo_frafico(col)
    plt.title(f"Años más frecuentes: {col}\nType: {df[col].dtype}")
    df[col].dt.year.value_counts().sort_index().plot(kind="bar")
    plt.show()
   

#----------------------------
# Selección de tipo de gráfico
#----------------------------

def elegir_grafico (df):
    """
    Detecta automáticamente el tipo de dato
    de cada columna del dataframe y selecciona
    el gráfico más adecuado.

    Reglas utilizadas:
    - Numéricas binarias → pastel
    - Numéricas con pocos valores → boxplot
    - Numéricas continuas → histograma
    - Categóricas binarias → pastel
    - Categóricas normales → barras
    - Booleanas → pastel
    - Fechas → barras por año

    Args:
        df (DataFrame):
            DataFrame a analizar y graficar.
    """

    for col in df.columns:

        if (df[col].dtype == "float" or  df[col].dtype == "int"):

            if df[col].nunique() == 2:

                pastel(df,col)

            elif df[col].nunique() < 10:

                grboxplot(df, col)  

            else:

                histograma(df, col) 

        elif df[col].dtype == "str":

            if df[col].nunique() == 2:

                pastel(df,col)

            else:            
                
                barres(df, col)

        elif df[col].dtype == "category":

            if df[col].nunique() == 2:

                pastel(df,col)

            else:
          
                barres(df, col)
        
        elif df[col].dtype == "bool":
            
            pastel(df,col)

        elif df[col].dtype in ('datetime', 'datetime64', 'datetime64[us]'):

            barres_any(df,col)
            
        else:
            
            print('otros tipos dtype de columna que no grafico', df[col].dtype)

        
#----------------------------
# Llamamos función
#----------------------------

elegir_grafico(df)
```

### Opción B

En la opción B al igual que en la opción A, trabajo sobre el df seleccionado con anterioridad en otra celda:

```python

# ==========================================
# FUNCIÓN PRINCIPAL
# ==========================================

def generar_graficos(df):

    # Recorremos todas las columnas
    for columna in df.columns:

        # Creamos figura nueva
        plt.figure(figsize=(8, 5))

        # Guardamos el dtype como string
        tipo_dato = str(df[columna].dtype)

        # Número de valores únicos
        cantidad_unicos = df[columna].nunique()

        # ==========================================
        # VARIABLES NUMÉRICAS
        # ==========================================

        if "int" in tipo_dato or "float" in tipo_dato:

            # --------------------------------------
            # NUMÉRICA BINARIA
            # Ejemplo: 0/1
            # --------------------------------------

            if cantidad_unicos == 2:

                df[columna].value_counts().sort_index().plot(kind="pie")

                plt.title(
                    f"Variable numérica binaria: {columna}\n(dtype: {tipo_dato})"
                )

                plt.xlabel(columna)
                plt.ylabel("Frecuencia")


            # --------------------------------------
            # NUMÉRICA CON POCOS VALORES
            # → Boxplot
            # --------------------------------------

            elif cantidad_unicos < 10:

                df[columna].plot(kind="box")

                plt.title(
                    f"Boxplot de la columna: {columna}\n(dtype: {tipo_dato})"
                )

                plt.ylabel(columna)


            # --------------------------------------
            # NUMÉRICA ESTÁNDAR
            # → Histograma
            # --------------------------------------

            else:

                df[columna].plot(kind="hist", bins=15)

                plt.title(
                    f"Histograma de la columna: {columna}\n(dtype: {tipo_dato})"
                )

                plt.xlabel(columna)
                plt.ylabel("Frecuencia")


        # ==========================================
        # VARIABLES CATEGÓRICAS / TEXTO
        # ==========================================

        elif (
            tipo_dato == "object"
            or tipo_dato == "category"
            or tipo_dato == "str"
        ):

            # --------------------------------------
            # CATEGÓRICA BINARIA
            # Ejemplo:
            # male/female
            # yes/no
            # --------------------------------------

            if cantidad_unicos == 2:

                df[columna].value_counts().plot(kind="pie", autopct="%1.1f%%")

                plt.title(
                    f"Variable categórica binaria: {columna}\n(dtype: {tipo_dato})"
                )

                plt.ylabel("")


            # --------------------------------------
            # CATEGÓRICA NORMAL
            # --------------------------------------

            else:

                df[columna].value_counts().head(10).plot(kind="bar")

                plt.title(
                    f"Gráfico de barras: {columna}\n(dtype: {tipo_dato})"
                )

                plt.xlabel(columna)
                plt.ylabel("Frecuencia")


        # ==========================================
        # VARIABLES BOOLEANAS
        # ==========================================

        elif tipo_dato == "bool":

            df[columna].value_counts().plot(
                kind="pie",
                autopct="%1.1f%%"
            )

            plt.title(
                f"Variable booleana: {columna}\n(dtype: {tipo_dato})"
            )

            plt.ylabel("")


        # ==========================================
        # VARIABLES FECHA
        # ==========================================

        elif "datetime" in tipo_dato:

            # Extraemos años
            años = df[columna].dt.year

            años.value_counts().sort_index().plot(kind="bar")

            plt.title(
                f"Años más frecuentes: {columna}\n(dtype: {tipo_dato})"
            )

            plt.xlabel("Año")
            plt.ylabel("Frecuencia")


        # ==========================================
        # OTROS TIPOS
        # ==========================================

        else:

            print(f"No se pudo generar gráfico para: {columna}")
            print(f"Tipo detectado: {tipo_dato}")

            continue


        

        # Mostrar gráfico
        plt.show()


# ==========================================
# EJECUCIÓN
# ==========================================

generar_graficos(df)
```

```python
import pandas as pd

opcion_a = "metodo del mejor vecino - Sin OT-R ni ciudad final - suma de rutas más cortas \n" \
            "entre ciudades para cada ciudad inicial. \n" \
            "finalmente se elije la ruta mas corta \n" \
            "- Tiempo total ejecución 0,1 seg.\n" \
            "- Total rutas analizadas:  13 rutas"

# ruta del fichoero a cargar (solo ruta sin nombre fichero)
ruta_file = "/Users/rogerdefez/Documents/Cursos i Llibres/BootCamp IT Academy/03_Especialitzacio/Sprint_10_Analitica_dades_Python/Datos originales/"

# carga fichero al dataframe
df = pd.read_excel(ruta_file + "matriu_distancies.xlsx", index_col=0)

# Borrado ciudades Palma y Las Palmas de Gran Canaria para ir en coche
matriz_ciudades = df.drop(["Palma","Las Palmas de Gran Canaria"]) # Borrado filas
matriz_ciudades = matriz_ciudades.drop(["Palma","Las Palmas de Gran Canaria"], axis=1)




# =====================================================
# FUNCIÓN PRINCIPAL
# =====================================================


def calcular_ruta(df_distancias, ciudad_inicio):

    """
    Calcula una ruta aproximada mínima utilizando
    el algoritmo del vecino más cercano.

    El algoritmo:
    - empieza en una ciudad inicial
    - busca siempre la ciudad no visitada más cercana
    - repite el proceso hasta visitar todas las ciudades

    Args:
        df_distancias (DataFrame):
            Matriz de distancias entre ciudades.

        ciudad_inicio (str):
            Ciudad desde la que comienza la ruta.

    Returns:
        tuple:
            - ciudades_visitadas (list):
                Lista ordenada con la ruta recorrida.

            - distancia_total (float):
                Distancia total recorrida en kilómetros.
    """

    # ---------------------------------------------
    # Lista de ciudades visitadas
    # ---------------------------------------------

    ciudades_visitadas = [ciudad_inicio]


    # ---------------------------------------------
    # Ciudad actual
    # ---------------------------------------------

    ciudad_actual = ciudad_inicio


    # ---------------------------------------------
    # Distancia total recorrida
    # ---------------------------------------------

    distancia_total = 0


    # ---------------------------------------------
    # Mientras queden ciudades por visitar
    # ---------------------------------------------

    while len(ciudades_visitadas) < len(df_distancias):

        # =========================================
        # Obtengo las distancias desde la ciudad actual
        # =========================================

        distancias = df_distancias.loc[ciudad_actual]


        # =========================================
        # Elimino ciudades ya visitadas
        # =========================================

        distancias_disponibles = distancias.drop(labels=ciudades_visitadas)


        # =========================================
        # Busco la ciudad más cercana
        # =========================================

        ciudad_mas_cercana = distancias_disponibles.idxmin()


        # =========================================
        # Distancia hasta esa ciudad
        # =========================================

        distancia_minima = distancias_disponibles.min()


        # =========================================
        # Actualizo distancia total
        # =========================================

        distancia_total += distancia_minima


        # =========================================
        # Añado ciudad a visitadas
        # =========================================

        ciudades_visitadas.append(ciudad_mas_cercana)


        # =========================================
        # La nueva ciudad actual será la seleccionada
        # =========================================

        ciudad_actual = ciudad_mas_cercana


    # =================================================
    # DEVOLVER RESULTADOS
    # =================================================

    return ciudades_visitadas, distancia_total


# =====================================================
# PROBAR DESDE TODAS LAS CIUDADES
# =====================================================


def encontrar_mejor_ruta(df_distancias):

    """
    Ejecuta el algoritmo del vecino más cercano
    empezando desde cada ciudad disponible.

    Compara todas las rutas generadas y guarda
    la de menor distancia total.

    Args:
        df_distancias (DataFrame):
            Matriz de distancias entre ciudades.

    Returns:
        tuple:
            - mejor_ruta (list):
                Ruta más corta encontrada.

            - mejor_distancia (float):
                Distancia total mínima encontrada.

            - mejor_ciudad_inicio (str):
                Ciudad inicial de la mejor ruta.
    """


    mejor_ruta = None

    mejor_distancia = float("inf")

    mejor_ciudad_inicio = None


    # =============================================
    # Repetimos el algoritmo empezando
    # desde cada ciudad
    # =============================================

    for ciudad in df_distancias.index:

        ruta, distancia = calcular_ruta(df_distancias, ciudad)


        print("=" * 60)
        print(f"Ciudad inicio: {ciudad}")
        print(f"Distancia total: {distancia:.0f} km")
        print(f"Ruta: {' -> '.join(ruta)}")
        print("=" * 60)
        print()


        # =========================================
        # Guardar mejor resultado
        # =========================================

        if distancia < mejor_distancia:

            mejor_distancia = distancia

            mejor_ruta = ruta

            mejor_ciudad_inicio = ciudad


    return mejor_ruta, mejor_distancia, mejor_ciudad_inicio


# =====================================================
# EJECUCIÓN FINAL
# =====================================================

ruta_final, distancia_final, ciudad_inicio_final = encontrar_mejor_ruta(matriz_ciudades)


print("\n" + "#" * 70)
print("MEJOR RUTA ENCONTRADA")
print("#" * 70)

print(f"\nCiudad inicial óptima aproximada: {ciudad_inicio_final}")

print(f"\nDistancia total aproximada: {distancia_final:.0f} km")

print(f"\nRuta final:")
print(" -> ".join(ruta_final))


opcion_a_ciudad = ciudad_inicio_final
opcion_a_ruta = ruta_final
opcion_a_distancia = distancia_final
```

```python
# Usando first_solution_strategy = .PATH_CHEAPEST_ARC - sin usar Ciudad_final

opcion_b1 = "Usando first_solution_strategy = .PATH_CHEAPEST_ARC - sin usar Ciudad_final \n" \
            "Siguiendo estrategia (first_solution_strategy) \n" \
            "por cada ciudad inicial encuentra la mejor ruta intentando menor tiempo \n" \
            "finalmente extrae y muestra la ruta mas corta de todas ellas\n" \
            "- Tiempo total ejecución 0,01 seg.\n" \
            "- Total rutas analizadas: 13 rutas mediante first_solution - ruta más corta "

#  ==========================================================
# IMPORTACIÓN DE LIBRERÍAS
# ==========================================================

import pandas as pd

from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp


# ==========================================================
# CARGA Y LIMPIEZA DEL DATAFRAME
# ==========================================================

def cargar_y_limpiar_datos(ruta_excel):
    """
    Carga el archivo Excel de distancias entre ciudades,
    elimina ciudades no accesibles en coche y limpia
    los datos para OR-Tools.

    Args:
        ruta_excel (str):
            Ruta completa del archivo Excel.

    Returns:
        tuple:
            - matriz_distancias (list)
            - ciudades (list)
    """

    # ---------------------------------------------
    # Cargar dataframe
    # ---------------------------------------------

    df = pd.read_excel(
        ruta_excel,
        index_col=0
    )

    # ---------------------------------------------
    # Eliminar ciudades no accesibles en coche
    # ---------------------------------------------

    ciudades_eliminar = [
        "Palma",
        "Las Palmas de Gran Canaria"
    ]

    df = df.drop(
        index=ciudades_eliminar,
        columns=ciudades_eliminar
    )

    # ---------------------------------------------
    # Limpiar valores
    # OR-Tools trabaja mejor con enteros puros
    # ---------------------------------------------

    df = df.fillna(0)

    df = df.astype(int)

    # ---------------------------------------------
    # Convertir dataframe a estructuras Python
    # ---------------------------------------------

    matriz_distancias = df.values.tolist()

    ciudades = df.index.tolist()

    return matriz_distancias, ciudades


# ==========================================================
# CREAR DATA MODEL
# ==========================================================

def crear_data_model(
    matriz_distancias,
    inicio
):
    """
    Crea el diccionario de datos para OR-Tools.

    Args:
        matriz_distancias (list):
            Matriz cuadrada de distancias.

        inicio (int):
            Índice ciudad inicio.

        final (int):
            Índice ciudad final.

    Returns:
        dict:
            Configuración del problema.
    """

    data = {

        "distance_matrix": matriz_distancias,

        "num_vehicles": 1,

        "depot": inicio

    }

    return data


# ==========================================================
# CALLBACK DE DISTANCIA
# ==========================================================

def crear_callback_distancia(
    data,
    manager
):
    """
    Crea la función callback que OR-Tools utiliza
    para calcular distancias entre nodos.

    Args:
        data (dict):
            Diccionario del modelo.

        manager:
            RoutingIndexManager de OR-Tools.

    Returns:
        function:
            Función callback.
    """

    def distance_callback(
        from_index,
        to_index
    ):

        from_node = manager.IndexToNode(
            from_index
        )

        to_node = manager.IndexToNode(
            to_index
        )

        return int(
            data["distance_matrix"]
            [from_node]
            [to_node]
        )

    return distance_callback


# ==========================================================
# EXTRAER RUTA Y DISTANCIA
# ==========================================================

def obtener_ruta_y_distancia(
    solution,
    routing,
    manager,
    ciudades
):
    """
    Extrae la ruta calculada y la distancia total.

    Args:
        solution:
            Solución OR-Tools.

        routing:
            RoutingModel.

        manager:
            RoutingIndexManager.

        ciudades (list):
            Lista nombres ciudades.

    Returns:
        tuple:
            - ruta (list)
            - distancia_total (int)
    """

    index = routing.Start(0)

    ruta = []

    distancia_total = 0

    while not routing.IsEnd(index):

        nodo = manager.IndexToNode(index)

        ruta.append(ciudades[nodo])

        previous_index = index

        index = solution.Value(
            routing.NextVar(index)
        )

        if not routing.IsEnd(index):

            distancia_total += (
                routing.GetArcCostForVehicle(
                    previous_index,
                    index,
                    0
                )
            )

    return ruta, distancia_total


# ==========================================================
# RESOLVER TSP (Travelling Salesman Problem)
# ==========================================================

def resolver_tsp(
    matriz_distancias,
    ciudades,
    ciudad_inicio
):
    """
    Resuelve el problema TSP para una combinación
    concreta de ciudad inicio y final.

    Args:
        matriz_distancias (list):
            Matriz de distancias.

        ciudades (list):
            Lista nombres ciudades.

        ciudad_inicio (str):
            Ciudad origen.

        ciudad_final (str):
            Ciudad destino final.

    Returns:
        tuple:
            - ruta (list)
            - distancia (int)
    """

    # ---------------------------------------------
    # Convertir nombres a índices
    # ---------------------------------------------

    inicio = ciudades.index(
        ciudad_inicio
    )


    # ---------------------------------------------
    # Crear modelo datos
    # ---------------------------------------------

    data = crear_data_model(
        matriz_distancias,
        inicio
    )

    # ---------------------------------------------
    # Crear manager
    # ---------------------------------------------

    manager = pywrapcp.RoutingIndexManager(
        len(data["distance_matrix"]),
        data["num_vehicles"],
        data["depot"]
    )

    # ---------------------------------------------
    # Crear routing model
    # ---------------------------------------------

    routing = pywrapcp.RoutingModel(
        manager
    )

    # ---------------------------------------------
    # Registrar callback distancias
    # ---------------------------------------------

    distance_callback = (
        crear_callback_distancia(
            data,
            manager
        )
    )

    transit_callback_index = (
        routing.RegisterTransitCallback(
            distance_callback
        )
    )

    routing.SetArcCostEvaluatorOfAllVehicles(
        transit_callback_index
    )

    # ---------------------------------------------
    # Parámetros búsqueda
    # ---------------------------------------------

    search_parameters = (
        pywrapcp.DefaultRoutingSearchParameters()
    )

    search_parameters.first_solution_strategy = (
        routing_enums_pb2
        .FirstSolutionStrategy
        .PATH_CHEAPEST_ARC
    )

    # ---------------------------------------------
    # Resolver problema
    # ---------------------------------------------

    solution = routing.SolveWithParameters(
        search_parameters
    )

    # ---------------------------------------------
    # Obtener resultados
    # ---------------------------------------------

    if solution:

        return obtener_ruta_y_distancia(
            solution,
            routing,
            manager,
            ciudades
        )

    return None, None


# ==========================================================
# BUSCAR MEJOR RUTA GLOBAL
# ==========================================================

def buscar_mejor_ruta(
    matriz_distancias,
    ciudades
):
    """
    Prueba todas las combinaciones posibles
    de inicio y final.

    Guarda la mejor ruta encontrada.

    Args:
        matriz_distancias (list):
            Matriz distancias.

        ciudades (list):
            Lista nombres ciudades.

    Returns:
        dict:
            Información mejor ruta.
    """

    mejor_resultado = {

        "ruta": None,

        "distancia": float("inf"),

        "inicio": None
    }

    # ---------------------------------------------
    # Probar todas las combinaciones
    # ---------------------------------------------

    for ciudad_inicio in ciudades:


        print("=" * 80)

        print(
            f"Probando ruta desde "
            f"{ciudad_inicio} "
        )

        ruta, distancia = resolver_tsp(
            matriz_distancias,
            ciudades,
            ciudad_inicio
            )

        if ruta:

            print(
                f"Distancia total: "
                f"{distancia} km"
            )

            print(
                " -> ".join(ruta)
            )

            # ---------------------------------
            # Guardar mejor resultado
            # ---------------------------------

            if distancia < mejor_resultado["distancia"]:

                mejor_resultado["ruta"] = ruta

                mejor_resultado["distancia"] = distancia

                mejor_resultado["inicio"] = ciudad_inicio


    return mejor_resultado


# ==========================================================
# FUNCIÓN PRINCIPAL ORQUESTADORA
# ==========================================================

def main():

    """
    Función principal de ejecución.
    """

    ruta_excel = ("/Users/rogerdefez/Documents/Cursos i Llibres/BootCamp IT Academy/03_Especialitzacio/Sprint_10_Analitica_dades_Python/Datos originales/")

    # ---------------------------------------------
    # Cargar datos
    # ---------------------------------------------

    matriz_distancias, ciudades = (cargar_y_limpiar_datos(ruta_excel+"matriu_distancies.xlsx"))

    # ---------------------------------------------
    # Buscar mejor ruta
    # ---------------------------------------------

    mejor_resultado = buscar_mejor_ruta(
        matriz_distancias,
        ciudades
    )

    # ---------------------------------------------
    # Mostrar mejor resultado
    # ---------------------------------------------

    print("\n" + "#" * 80)

    print("MEJOR RUTA ENCONTRADA")

    print("#" * 80)

    print(
        f"\nCiudad inicio: "
        f"{mejor_resultado['inicio']}"
    )


    print(
        f"\nDistancia mínima: "
        f"{mejor_resultado['distancia']} km"
    )

    print("\nRuta final:")

    print(
        " -> ".join(
            mejor_resultado["ruta"]
        )
    )

    return mejor_resultado

# ==========================================================
# EJECUCIÓN APP
# ==========================================================


mejor_resultado = main()

opcion_b1_ciudad = mejor_resultado['inicio']
opcion_b1_ruta = mejor_resultado["ruta"]
opcion_b1_distancia = mejor_resultado['distancia']
```

```python
# Importamos las librerias

import pandas as pd

from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp


# ---------------------------------
# CARGO DATAFRAME
# ---------------------------------

ruta_file = "/Users/rogerdefez/Documents/Cursos i Llibres/BootCamp IT Academy/03_Especialitzacio/Sprint_10_Analitica_dades_Python/Datos originales/"

df = pd.read_excel(
    ruta_file + "matriu_distancies.xlsx",
    index_col=0
)


# ---------------------------------
# ELIMINO ISLAS
# ---------------------------------
ciudades_eliminar = [
    "Palma",
    "Las Palmas de Gran Canaria"
]

df = df.drop(index=ciudades_eliminar)
df = df.drop(columns=ciudades_eliminar)

# importante convierto los valores del df a integer y limpio 0 ya que por defecto al cargarlos del excel me entran como float 
# y resulta que OR-Tools Routing es muy sensible a tipos, enteros y matrices limpias

df = df.fillna(0)
df = df.astype(int)  

# ---------------------------------
# MATRIZ Y LISTA CIUDADES
# ---------------------------------

# separo el dataframe en los valores para alimentar la "distance_matrix" y 
# la lista del indice de ciudades para convertir resultados al final

matriz_distancias = df.values.tolist()

ciudades = df.index.tolist()


# ---------------------------------
# CREAR DATA MODEL
# ---------------------------------

def crear_data_model(inicio, final):

    data = {}

    data["distance_matrix"] = matriz_distancias

    data["num_vehicles"] = 1

    data["depot"] = inicio

    data["depot_final"] = final

    return data


# ---------------------------------
# IMPRIMIR SOLUCIÓN
# ---------------------------------

def obtener_ruta_y_distancia(
    solution,
    routing,
    manager,
    ciudades
):

    index = routing.Start(0)

    ruta = []

    distancia_total = 0


    while not routing.IsEnd(index):

        nodo = manager.IndexToNode(index)

        ruta.append(ciudades[nodo])

        previous_index = index

        index = solution.Value(
            routing.NextVar(index)
        )

        distancia_total += routing.GetArcCostForVehicle(
            previous_index,
            index,
            0
        )


    nodo_final = manager.IndexToNode(index)

    ruta.append(ciudades[nodo_final])

    return ruta, distancia_total


# ==========================================================
# RESOLVER TSP DESDE UNA CIUDAD
# ==========================================================

def resolver_tsp(ciudad_inicio, ciudad_final):

    deposito = ciudades.index(ciudad_inicio)

    deposito_final = ciudades.index(ciudad_final)

    data = crear_data_model(deposito, deposito_final)


    # ------------------------------------------------------
    # MANAGER
    # ------------------------------------------------------

    manager = pywrapcp.RoutingIndexManager(len(data["distance_matrix"]),
        data["num_vehicles"],
        [data["depot"]],
        [data["depot_final"]]
)


    # ------------------------------------------------------
    # ROUTING
    # ------------------------------------------------------

    routing = pywrapcp.RoutingModel(manager)


    # ------------------------------------------------------
    # CALLBACK DISTANCIAS
    # ------------------------------------------------------

    def distance_callback(from_index, to_index):

        from_node = manager.IndexToNode(from_index)

        to_node = manager.IndexToNode(to_index)

        return int(data["distance_matrix"][from_node][to_node])


    transit_callback_index = routing.RegisterTransitCallback(
        distance_callback
    )


    routing.SetArcCostEvaluatorOfAllVehicles(
        transit_callback_index
    )


    # ------------------------------------------------------
    # PARÁMETROS BÚSQUEDA
    # ------------------------------------------------------

    search_parameters = (
        pywrapcp.DefaultRoutingSearchParameters()
    )

    search_parameters.first_solution_strategy = (
        routing_enums_pb2.FirstSolutionStrategy.PATH_CHEAPEST_ARC
    )


    # ------------------------------------------------------
    # RESOLVER
    # ------------------------------------------------------

    solution = routing.SolveWithParameters(
        search_parameters
    )


    if solution:

        ruta, distancia = obtener_ruta_y_distancia(
            solution,
            routing,
            manager,
            ciudades
        )

        return ruta, distancia


    return None, None


# ==========================================================
# PROBAR TODAS LAS COMBINACIONES
# INICIO / FINAL
# ==========================================================

mejor_ruta = None

mejor_distancia = float("inf")

mejor_ciudad_inicio = None

mejor_ciudad_final = None


for ciudad_inicio in ciudades:

    for ciudad_final in ciudades:

        # Evitar inicio y final iguales
        if ciudad_inicio != ciudad_final:

            print("=" * 80)

            print(
                f"Probando ruta desde "
                f"{ciudad_inicio} hasta {ciudad_final}"
            )

            ruta, distancia = resolver_tsp(
                ciudad_inicio,
                ciudad_final
            )


            if ruta:

                print(f"Distancia total: {distancia} km")

                print("Ruta:")

                print(" -> ".join(ruta))


                # ======================================
                # GUARDAR MEJOR RESULTADO
                # ======================================

                if distancia < mejor_distancia:

                    mejor_distancia = distancia

                    mejor_ruta = ruta

                    mejor_ciudad_inicio = ciudad_inicio

                    mejor_ciudad_final = ciudad_final


            print()


# ==========================================================
# MEJOR RESULTADO FINAL
# ==========================================================

print("\n" + "#" * 80)

print("MEJOR RUTA ENCONTRADA")

print("#" * 80)

print(f"\nCiudad inicio: {mejor_ciudad_inicio}")

print(f"Ciudad final: {mejor_ciudad_final}")

print(f"\nDistancia mínima: {mejor_distancia} km")

print("\nRuta final:")

print(" -> ".join(mejor_ruta))
```
